```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["font.size"] = 11

# ============================================================
# 2. FILE PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Sugar Industry Performance Analysis and Production Efficiency"
)

DATASET_PATH = BASE_DIR / (
    "performance_of_sugar_industry_sugar_season_2019.csv"
)

H5_PATH = BASE_DIR / "sugar_production_model.h5"
PKL_PATH = BASE_DIR / "sugar_production_model.pkl"
YAML_PATH = BASE_DIR / "config.yaml"
JSON_PATH = BASE_DIR / "metrics.json"

BASE_DIR.mkdir(parents=True, exist_ok=True)

required_files = [
    DATASET_PATH,
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]

for file_path in required_files:

    if not file_path.exists():

        raise FileNotFoundError(
            f"Required file missing:\n{file_path}"
        )

print("=" * 75)
print("SUGAR INDUSTRY PERFORMANCE ANALYSIS")
print("MODEL EVALUATION AND VISUALIZATION")
print("=" * 75)

# ============================================================
# 3. LOAD CONFIGURATION AND METRICS
# ============================================================

with open(YAML_PATH, "r", encoding="utf-8") as file:

    config = yaml.safe_load(file)

with open(JSON_PATH, "r", encoding="utf-8") as file:

    metrics = json.load(file)

FEATURES = config["dataset"]["features"]

TARGET = config["dataset"]["target"]

print("\nFeatures:", FEATURES)
print("Target:", TARGET)

# ============================================================
# 4. LOAD TRAINED MODELS
# ============================================================

rf_model = joblib.load(PKL_PATH)

ann_model = tf.keras.models.load_model(
    str(H5_PATH),
    compile=False
)

print("\nRandom Forest model loaded successfully.")
print("ANN model loaded successfully.")

# ============================================================
# 5. LOAD ORIGINAL DATASET
# ============================================================

raw_df = pd.read_csv(DATASET_PATH)

raw_df.columns = raw_df.columns.str.strip()

raw_df["Parameters"] = (
    raw_df["Parameters"]
    .astype(str)
    .str.strip()
)

print("\nOriginal Dataset Shape:", raw_df.shape)

print("\nOriginal Dataset:")
print(raw_df.to_string(index=False))

# ============================================================
# 6. DATA TRANSFORMATION
# ============================================================

parameter_mapping = {

    "Number of Factories Working":
        "working_factories",

    "Installed Crushing Capacity (in Lakh Tonnes) per day":
        "installed_crushing_capacity",

    "Cane Crushed (in Lakh M.T.)":
        "cane_crushed",

    "Sugar Production from Cane (in Lakh M.T.)":
        "sugar_production",

    "Recovery Rate (in Percentage)":
        "recovery_rate"
}

df = raw_df.set_index("Parameters")

df = df.drop(
    columns=["S.No"],
    errors="ignore"
)

df = df.rename(index=parameter_mapping)

df = df.T.reset_index()

df = df.rename(
    columns={"index": "region_season"}
)

df.columns.name = None

# ============================================================
# 7. EXTRACT REGION AND SEASON
# ============================================================

df["region"] = np.where(

    df["region_season"].str.contains(
        "Tamil Nadu",
        case=False
    ),

    "Tamil Nadu",

    "All India"
)

df["season"] = df["region_season"].str.extract(
    r"(\d{4}-\d{2})"
)

df["year"] = pd.to_numeric(
    df["season"].str[:4],
    errors="coerce"
)

df["provisional"] = (
    df["region_season"]
    .str.contains(
        "Provisional",
        case=False
    )
)

df["region_code"] = df["region"].map({

    "Tamil Nadu": 0,

    "All India": 1
})

numeric_columns = [
    "working_factories",
    "installed_crushing_capacity",
    "cane_crushed",
    "sugar_production",
    "recovery_rate"
]

for column in numeric_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

df = df.dropna(
    subset=FEATURES + [TARGET]
).reset_index(drop=True)

df["label"] = (
    df["region"] +
    "\n" +
    df["season"]
)

print("\nProcessed Dataset Shape:", df.shape)

print("\nProcessed Dataset:")
print(df.to_string(index=False))

# ============================================================
# 8. FEATURE ENGINEERING
# ============================================================

df["production_per_factory"] = (

    df["sugar_production"] /
    df["working_factories"]
)

df["cane_per_factory"] = (

    df["cane_crushed"] /
    df["working_factories"]
)

df["effective_recovery_rate"] = (

    df["sugar_production"] /
    df["cane_crushed"]

) * 100

df["recovery_gap"] = (

    df["effective_recovery_rate"] -
    df["recovery_rate"]
)

df["mathematical_prediction"] = (

    df["cane_crushed"] *
    df["recovery_rate"] / 100
)

# ============================================================
# 9. GENERATE MODEL PREDICTIONS
# ============================================================

X = df[FEATURES].astype("float32")

y = df[TARGET].to_numpy(dtype=float)

rf_predictions = rf_model.predict(X)

ann_predictions = ann_model.predict(

    X.to_numpy(dtype=np.float32),

    verbose=0

).flatten()

baseline_predictions = (
    df["mathematical_prediction"].to_numpy()
)

df["random_forest_prediction"] = rf_predictions

df["ann_prediction"] = ann_predictions

# ============================================================
# 10. EVALUATION FUNCTION
# ============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(
        actual,
        dtype=float
    ).flatten()

    predicted = np.asarray(
        predicted,
        dtype=float
    ).flatten()

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    nonzero = actual != 0

    if np.any(nonzero):

        mape = np.mean(

            np.abs(

                (
                    actual[nonzero] -
                    predicted[nonzero]
                ) /
                actual[nonzero]

            )

        ) * 100

        score = 100 - mape

    else:

        mape = np.nan
        score = np.nan

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2_Score": float(r2),

        "MAPE_Percentage": float(mape),

        "Estimation_Score_Percentage": float(score)
    }

# ============================================================
# 11. MODEL EVALUATION
# ============================================================

model_results = {

    "Mathematical Baseline": evaluate_model(

        y,
        baseline_predictions
    ),

    "Random Forest": evaluate_model(

        y,
        rf_predictions
    ),

    "ANN": evaluate_model(

        y,
        ann_predictions
    )
}

comparison_df = pd.DataFrame(
    model_results
).T

comparison_df.index.name = "Model"

comparison_df.to_csv(

    BASE_DIR / "model_comparison.csv",

    index=True
)

print("\n" + "=" * 75)
print("MODEL PERFORMANCE ON ORIGINAL OBSERVATIONS")
print("=" * 75)

print(comparison_df.round(4))

# ============================================================
# 12. GRAPH SAVING FUNCTION
# ============================================================

def save_and_show(filename):

    plt.tight_layout()

    plt.savefig(

        BASE_DIR / filename,

        dpi=300,

        bbox_inches="tight",

        facecolor="white"
    )

    plt.show()

    plt.close()

    print(f"[SAVED] {filename}")

# ============================================================
# 13. ACCURACY GRAPH
# ============================================================

# This is a regression problem.
# Estimation Score = 100 - MAPE.
# It is not classification accuracy.

simulation_metrics = metrics[
    "simulation_test_metrics"
]

model_names = [
    "Mathematical Baseline",
    "Random Forest",
    "ANN"
]

simulation_scores = [

    simulation_metrics[
        "Mathematical_Baseline"
    ]["Estimation_Score_Percentage"],

    simulation_metrics[
        "Random_Forest"
    ]["Estimation_Score_Percentage"],

    simulation_metrics[
        "Artificial_Neural_Network"
    ]["Estimation_Score_Percentage"]
]

original_scores = [

    model_results[name][
        "Estimation_Score_Percentage"
    ]

    for name in model_names
]

positions = np.arange(len(model_names))

width = 0.35

plt.figure(figsize=(12, 7))

bars1 = plt.bar(

    positions - width / 2,

    simulation_scores,

    width,

    label="Simulated Test Data",

    color="#3498db",

    edgecolor="black"
)

bars2 = plt.bar(

    positions + width / 2,

    original_scores,

    width,

    label="Original Observations",

    color="#2ecc71",

    edgecolor="black"
)

for bars in [bars1, bars2]:

    for bar in bars:

        height = bar.get_height()

        plt.annotate(

            f"{height:.2f}%",

            (
                bar.get_x() +
                bar.get_width() / 2,

                height
            ),

            xytext=(0, 5),

            textcoords="offset points",

            ha="center",

            fontsize=10,

            fontweight="bold"
        )

plt.xticks(
    positions,
    model_names
)

plt.ylabel("Estimation Score (%)")

plt.xlabel("Algorithms")

plt.title(

    "Sugar Production Estimation Performance\n"
    "(100 - MAPE; Not Classification Accuracy)",

    fontsize=16,

    fontweight="bold"
)

plt.ylim(

    min(0, min(original_scores + simulation_scores) - 10),

    max(110, max(original_scores + simulation_scores) + 10)
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

save_and_show("accuracy_graph.png")

# ============================================================
# 14. CORRELATION HEATMAP
# ============================================================

heatmap_columns = [

    "working_factories",

    "installed_crushing_capacity",

    "cane_crushed",

    "sugar_production",

    "recovery_rate",

    "production_per_factory",

    "cane_per_factory",

    "effective_recovery_rate"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(13, 10))

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".2f",

    cmap="coolwarm",

    center=0,

    vmin=-1,

    vmax=1,

    linewidths=0.5,

    square=True,

    cbar_kws={
        "label": "Correlation Coefficient"
    }
)

plt.title(

    "Sugar Industry Feature Correlation Heatmap\n"
    "Descriptive Analysis of Four Observations",

    fontsize=16,

    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(rotation=0)

save_and_show("heatmap.png")

# ============================================================
# 15. MODEL COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(15, 11)
)

comparison_metrics = [

    "MAE",

    "RMSE",

    "R2_Score",

    "MAPE_Percentage"
]

metric_titles = [

    "Mean Absolute Error",

    "Root Mean Squared Error",

    "R² Score",

    "Mean Absolute Percentage Error"
]

colors = [

    "#3498db",

    "#2ecc71",

    "#e67e22"
]

for ax, metric, title in zip(

    axes.flatten(),

    comparison_metrics,

    metric_titles
):

    values = comparison_df[metric]

    bars = ax.bar(

        comparison_df.index,

        values,

        color=colors,

        edgecolor="black"
    )

    ax.set_title(

        title,

        fontsize=13,

        fontweight="bold"
    )

    ax.set_ylabel(metric)

    ax.tick_params(
        axis="x",
        rotation=15
    )

    ax.grid(
        axis="y",
        alpha=0.3
    )

    ax.margins(y=0.18)

    for bar in bars:

        height = bar.get_height()

        ax.annotate(

            f"{height:.4f}",

            (
                bar.get_x() +
                bar.get_width() / 2,

                height
            ),

            xytext=(0, 4),

            textcoords="offset points",

            ha="center",

            fontsize=10
        )

fig.suptitle(

    "Comparative Analysis of Sugar Production Models\n"
    "Original Region-Season Observations",

    fontsize=18,

    fontweight="bold"
)

fig.tight_layout(
    rect=[0, 0, 1, 0.94]
)

save_and_show("comparison_graph.png")

# ============================================================
# 16. PRODUCTION EFFICIENCY ANALYSIS
# ============================================================

# Simulate a 5% relative improvement in recovery rate.

RECOVERY_IMPROVEMENT = 0.05

MAX_RECOVERY_RATE = 13.0

df["improved_recovery_rate"] = np.minimum(

    df["recovery_rate"] *
    (1 + RECOVERY_IMPROVEMENT),

    MAX_RECOVERY_RATE
)

df["improved_sugar_production"] = (

    df["cane_crushed"] *
    df["improved_recovery_rate"] / 100
)

df["additional_sugar_production"] = (

    df["improved_sugar_production"] -
    df["mathematical_prediction"]
)

df["production_improvement_percentage"] = (

    df["additional_sugar_production"] /
    df["mathematical_prediction"]

) * 100

df["rf_absolute_error"] = np.abs(

    df["sugar_production"] -
    df["random_forest_prediction"]
)

df["ann_absolute_error"] = np.abs(

    df["sugar_production"] -
    df["ann_prediction"]
)

df["baseline_absolute_error"] = np.abs(

    df["sugar_production"] -
    df["mathematical_prediction"]
)

# ============================================================
# 17. SAVE RESULT CSV
# ============================================================

result_columns = [

    "region",

    "season",

    "provisional",

    "working_factories",

    "installed_crushing_capacity",

    "cane_crushed",

    "sugar_production",

    "recovery_rate",

    "effective_recovery_rate",

    "production_per_factory",

    "cane_per_factory",

    "recovery_gap",

    "mathematical_prediction",

    "random_forest_prediction",

    "ann_prediction",

    "baseline_absolute_error",

    "rf_absolute_error",

    "ann_absolute_error",

    "improved_recovery_rate",

    "improved_sugar_production",

    "additional_sugar_production",

    "production_improvement_percentage"
]

result_df = df[result_columns].copy()

result_df = result_df.round(6)

result_df.to_csv(

    BASE_DIR / "result.csv",

    index=False
)

print("\n" + "=" * 75)
print("SUGAR INDUSTRY PERFORMANCE RESULTS")
print("=" * 75)

print(result_df.to_string(index=False))

print("\n[SAVED] result.csv")

# ============================================================
# 18. RESULT GRAPH
# ============================================================

# Separate regions because their production scales
# differ substantially.

regions = df["region"].unique()

fig, axes = plt.subplots(

    1,

    len(regions),

    figsize=(16, 7),

    squeeze=False
)

for ax, region in zip(

    axes.flatten(),

    regions
):

    region_df = df[
        df["region"] == region
    ].sort_values("year")

    positions = np.arange(len(region_df))

    width = 0.16

    graph_columns = [

        "sugar_production",

        "mathematical_prediction",

        "random_forest_prediction",

        "ann_prediction",

        "improved_sugar_production"
    ]

    graph_labels = [

        "Reported Production",

        "Mathematical Baseline",

        "Random Forest",

        "ANN",

        "Improved Scenario"
    ]

    graph_colors = [

        "#34495e",

        "#3498db",

        "#2ecc71",

        "#e67e22",

        "#9b59b6"
    ]

    for i, (column, label, color) in enumerate(

        zip(
            graph_columns,
            graph_labels,
            graph_colors
        )
    ):

        bars = ax.bar(

            positions + (i - 2) * width,

            region_df[column],

            width,

            label=label,

            color=color,

            edgecolor="black"
        )

        for bar in bars:

            height = bar.get_height()

            ax.annotate(

                f"{height:.2f}",

                (
                    bar.get_x() +
                    bar.get_width() / 2,

                    height
                ),

                xytext=(0, 4),

                textcoords="offset points",

                ha="center",

                rotation=90,

                fontsize=8
            )

    ax.set_xticks(positions)

    ax.set_xticklabels(
        region_df["season"]
    )

    ax.set_title(

        region,

        fontsize=14,

        fontweight="bold"
    )

    ax.set_ylabel(
        "Sugar Production (Lakh MT)"
    )

    ax.set_xlabel("Sugar Season")

    ax.margins(y=0.25)

    ax.legend(fontsize=8)

    ax.grid(
        axis="y",
        alpha=0.3
    )

fig.suptitle(

    "Sugar Industry Performance and Efficiency Analysis",

    fontsize=18,

    fontweight="bold"
)

save_and_show("result_graph.png")

# ============================================================
# 19. GENERATE PREDICTION SCENARIOS
# ============================================================

# Use the most recent available observation from
# each region as the reference operating condition.
#
# These are hypothetical scenarios, not future
# year forecasts.

latest_df = (

    df.sort_values("year")

    .groupby("region")

    .tail(1)

    .reset_index(drop=True)
)

scenarios = [

    {
        "scenario": "Current Conditions",
        "factory_change": 0.00,
        "cane_change": 0.00,
        "recovery_change": 0.00
    },

    {
        "scenario": "Recovery Improvement",
        "factory_change": 0.00,
        "cane_change": 0.00,
        "recovery_change": 0.05
    },

    {
        "scenario": "Increased Cane Availability",
        "factory_change": 0.00,
        "cane_change": 0.10,
        "recovery_change": 0.00
    },

    {
        "scenario": "Factory Expansion",
        "factory_change": 0.05,
        "cane_change": 0.05,
        "recovery_change": 0.00
    },

    {
        "scenario": "Combined Improvement",
        "factory_change": 0.05,
        "cane_change": 0.10,
        "recovery_change": 0.05
    },

    {
        "scenario": "Reduced Cane Availability",
        "factory_change": 0.00,
        "cane_change": -0.10,
        "recovery_change": 0.00
    }
]

prediction_records = []

for _, reference in latest_df.iterrows():

    for scenario in scenarios:

        factories = max(

            1,

            int(

                round(

                    reference["working_factories"] *

                    (1 + scenario["factory_change"])
                )
            )
        )

        cane_crushed = (

            reference["cane_crushed"] *

            (1 + scenario["cane_change"])
        )

        recovery_rate = np.clip(

            reference["recovery_rate"] *

            (1 + scenario["recovery_change"]),

            6.0,

            MAX_RECOVERY_RATE
        )

        prediction_records.append({

            "region": reference["region"],

            "reference_season": reference["season"],

            "scenario": scenario["scenario"],

            "working_factories": factories,

            "cane_crushed": float(cane_crushed),

            "recovery_rate": float(recovery_rate),

            "reference_reported_production": float(
                reference["sugar_production"]
            ),

            "reference_calculated_production": float(
                reference["mathematical_prediction"]
            ),

            "data_type": "Hypothetical Simulation"
        })

prediction_df = pd.DataFrame(
    prediction_records
)

# ============================================================
# 20. PREDICT SUGAR PRODUCTION
# ============================================================

X_future = prediction_df[
    FEATURES
].astype("float32")

prediction_df["mathematical_prediction"] = (

    prediction_df["cane_crushed"] *

    prediction_df["recovery_rate"] / 100
)

prediction_df["random_forest_prediction"] = (

    rf_model.predict(X_future)
)

prediction_df["ann_prediction"] = (

    ann_model.predict(

        X_future.to_numpy(dtype=np.float32),

        verbose=0

    ).flatten()
)

prediction_df["ensemble_prediction"] = (

    prediction_df["random_forest_prediction"] +

    prediction_df["ann_prediction"]

) / 2

# Compare simulated production with the mathematical
# reference to avoid mixing formula discrepancies
# with hypothetical efficiency gains.

prediction_df["expected_production_change"] = (

    prediction_df["mathematical_prediction"] -

    prediction_df["reference_calculated_production"]
)

prediction_df["expected_improvement_percentage"] = (

    prediction_df["expected_production_change"] /

    prediction_df["reference_calculated_production"]

) * 100

# ============================================================
# 21. SCENARIO CLASSIFICATION
# ============================================================

prediction_df["performance_category"] = np.select(

    [

        prediction_df[
            "expected_improvement_percentage"
        ] >= 10,

        prediction_df[
            "expected_improvement_percentage"
        ] > 0,

        prediction_df[
            "expected_improvement_percentage"
        ] < 0
    ],

    [

        "High Improvement",

        "Moderate Improvement",

        "Production Decline"
    ],

    default="Stable"
)

# ============================================================
# 22. SAVE PREDICTION CSV
# ============================================================

prediction_df = prediction_df.round(6)

prediction_df.to_csv(

    BASE_DIR / "prediction.csv",

    index=False
)

print("\n" + "=" * 75)
print("SUGAR PRODUCTION SCENARIO PREDICTIONS")
print("=" * 75)

print(prediction_df.to_string(index=False))

print("\n[SAVED] prediction.csv")

# ============================================================
# 23. PREDICTION GRAPH
# ============================================================

fig, axes = plt.subplots(

    2,

    1,

    figsize=(16, 13)
)

for ax, region in zip(

    axes.flatten(),

    prediction_df["region"].unique()
):

    region_predictions = prediction_df[

        prediction_df["region"] == region

    ].reset_index(drop=True)

    positions = np.arange(
        len(region_predictions)
    )

    ax.plot(

        positions,

        region_predictions[
            "mathematical_prediction"
        ],

        marker="o",

        linewidth=2.5,

        color="#3498db",

        label="Mathematical Baseline"
    )

    ax.plot(

        positions,

        region_predictions[
            "random_forest_prediction"
        ],

        marker="s",

        linewidth=2.5,

        color="#2ecc71",

        label="Random Forest"
    )

    ax.plot(

        positions,

        region_predictions[
            "ann_prediction"
        ],

        marker="^",

        linewidth=2.5,

        color="#e67e22",

        label="ANN"
    )

    ax.plot(

        positions,

        region_predictions[
            "ensemble_prediction"
        ],

        marker="D",

        linewidth=2.5,

        linestyle="--",

        color="#9b59b6",

        label="Ensemble Prediction"
    )

    ax.axhline(

        y=region_predictions[
            "reference_reported_production"
        ].iloc[0],

        color="red",

        linestyle=":",

        linewidth=2,

        label="Reported Reference Production"
    )

    ax.set_xticks(positions)

    ax.set_xticklabels(

        region_predictions["scenario"],

        rotation=15,

        ha="right"
    )

    ax.set_title(

        f"{region} - Sugar Production Scenarios",

        fontsize=14,

        fontweight="bold"
    )

    ax.set_ylabel(

        "Estimated Sugar Production (Lakh MT)"
    )

    ax.legend()

    ax.grid(alpha=0.3)

fig.suptitle(

    "AI-Based Sugar Production Scenario Predictions",

    fontsize=18,

    fontweight="bold"
)

save_and_show("prediction_graph.png")

# ============================================================
# 24. ANN TRAINING LOSS GRAPH
# ============================================================

training_history = metrics.get(

    "training_history",

    {}
)

if (

    "loss" in training_history and

    "val_loss" in training_history
):

    training_loss = training_history["loss"]

    validation_loss = training_history["val_loss"]

    epochs = np.arange(

        1,

        len(training_loss) + 1
    )

    plt.figure(figsize=(12, 7))

    plt.plot(

        epochs,

        training_loss,

        label="Training Loss",

        linewidth=2.5,

        color="#3498db"
    )

    plt.plot(

        epochs,

        validation_loss,

        label="Validation Loss",

        linewidth=2.5,

        color="#e74c3c"
    )

    best_epoch = np.argmin(
        validation_loss
    ) + 1

    plt.axvline(

        best_epoch,

        color="green",

        linestyle="--",

        label=f"Best Epoch: {best_epoch}"
    )

    plt.title(

        "ANN Training and Validation Loss\n"
        "Simulated Training Dataset",

        fontsize=16,

        fontweight="bold"
    )

    plt.xlabel("Epoch")

    plt.ylabel("Mean Squared Error")

    plt.yscale("symlog", linthresh=1e-6)

    plt.legend()

    plt.grid(alpha=0.3)

    save_and_show("training_loss_graph.png")

else:

    print("\nTraining history is unavailable.")

# ============================================================
# 25. FEATURE IMPORTANCE GRAPH
# ============================================================

feature_importance = metrics.get(

    "feature_importance",

    {}
)

if feature_importance:

    importance_df = pd.DataFrame({

        "Feature": list(
            feature_importance.keys()
        ),

        "Importance": list(
            feature_importance.values()
        )
    })

    importance_df = importance_df.sort_values(

        "Importance",

        ascending=True
    )

    plt.figure(figsize=(11, 6))

    bars = plt.barh(

        importance_df["Feature"],

        importance_df["Importance"],

        color="#2980b9",

        edgecolor="black"
    )

    for bar in bars:

        value = bar.get_width()

        plt.text(

            value + 0.01,

            bar.get_y() +
            bar.get_height() / 2,

            f"{value:.4f}",

            va="center",

            fontweight="bold"
        )

    plt.title(

        "Random Forest Feature Importance\n"
        "Model Trained on Simulated Data",

        fontsize=16,

        fontweight="bold"
    )

    plt.xlabel("Feature Importance")

    plt.xlim(

        0,

        max(
            1.0,
            importance_df["Importance"].max() + 0.15
        )
    )

    plt.grid(

        axis="x",

        alpha=0.3
    )

    save_and_show(
        "feature_importance_graph.png"
    )

# ============================================================
# 26. BEST MODEL IDENTIFICATION
# ============================================================

best_model_rmse = comparison_df[
    "RMSE"
].idxmin()

best_model_r2 = comparison_df[
    "R2_Score"
].idxmax()

best_model_mape = comparison_df[
    "MAPE_Percentage"
].idxmin()

print("\n" + "=" * 75)
print("MODEL PERFORMANCE SUMMARY")
print("=" * 75)

print(

    f"\nLowest RMSE Model: "
    f"{best_model_rmse}"
)

print(

    f"RMSE: "
    f"{comparison_df.loc[best_model_rmse, 'RMSE']:.6f}"
)

print(

    f"\nHighest R2 Model: "
    f"{best_model_r2}"
)

print(

    f"R2 Score: "
    f"{comparison_df.loc[best_model_r2, 'R2_Score']:.6f}"
)

print(

    f"\nLowest MAPE Model: "
    f"{best_model_mape}"
)

print(

    f"MAPE: "
    f"{comparison_df.loc[best_model_mape, 'MAPE_Percentage']:.4f}%"
)

# ============================================================
# 27. BEST PRODUCTION SCENARIO
# ============================================================

print("\n" + "=" * 75)
print("BEST HYPOTHETICAL PRODUCTION SCENARIOS")
print("=" * 75)

for region in prediction_df["region"].unique():

    regional_predictions = prediction_df[

        prediction_df["region"] == region
    ]

    best_index = regional_predictions[
        "mathematical_prediction"
    ].idxmax()

    best_scenario = prediction_df.loc[
        best_index
    ]

    print(f"\nRegion: {region}")

    print(

        "Best Scenario:",

        best_scenario["scenario"]
    )

    print(

        f"Working Factories: "
        f"{best_scenario['working_factories']}"
    )

    print(

        f"Cane Crushed: "
        f"{best_scenario['cane_crushed']:.2f} Lakh MT"
    )

    print(

        f"Recovery Rate: "
        f"{best_scenario['recovery_rate']:.2f}%"
    )

    print(

        f"Estimated Production: "
        f"{best_scenario['mathematical_prediction']:.4f} "
        f"Lakh MT"
    )

    print(

        f"Expected Improvement: "
        f"{best_scenario['expected_improvement_percentage']:.2f}%"
    )

    print(

        f"Performance Category: "
        f"{best_scenario['performance_category']}"
    )

# ============================================================
# 28. ADDITIONAL REGIONAL PERFORMANCE SUMMARY
# ============================================================

regional_summary = df.groupby("region").agg(

    Average_Working_Factories=(
        "working_factories",
        "mean"
    ),

    Average_Cane_Crushed=(
        "cane_crushed",
        "mean"
    ),

    Average_Sugar_Production=(
        "sugar_production",
        "mean"
    ),

    Average_Recovery_Rate=(
        "recovery_rate",
        "mean"
    ),

    Average_Production_Per_Factory=(
        "production_per_factory",
        "mean"
    ),

    Average_Improved_Production=(
        "improved_sugar_production",
        "mean"
    )
)

regional_summary.to_csv(

    BASE_DIR / "regional_summary.csv"
)

print("\nRegional Performance Summary:")

print(regional_summary.round(4))

print("\n[SAVED] regional_summary.csv")

# ============================================================
# 29. FINAL FILE VERIFICATION
# ============================================================

output_files = [

    "accuracy_graph.png",

    "heatmap.png",

    "comparison_graph.png",

    "result.csv",

    "result_graph.png",

    "prediction.csv",

    "prediction_graph.png",

    "training_loss_graph.png",

    "feature_importance_graph.png",

    "model_comparison.csv",

    "regional_summary.csv"
]

print("\n" + "=" * 75)
print("GENERATED OUTPUT FILES")
print("=" * 75)

success_count = 0

for filename in output_files:

    file_path = BASE_DIR / filename

    if file_path.exists():

        size_kb = (

            file_path.stat().st_size / 1024
        )

        print(

            f"[SUCCESS] {filename} "

            f"({size_kb:.2f} KB)"
        )

        success_count += 1

    else:

        print(f"[NOT GENERATED] {filename}")

print(

    f"\nTotal Generated Files: "

    f"{success_count}/{len(output_files)}"
)

print("\nOutput Directory:")

print(BASE_DIR)

print("\n" + "=" * 75)
print("IMPORTANT DATA LIMITATIONS")
print("=" * 75)

print(

    "1. The original dataset contains only four "
    "region-season observations."
)

print(

    "2. Random Forest and ANN were trained on "
    "mathematically simulated data."
)

print(

    "3. Estimation score is 100 minus MAPE, "
    "not classification accuracy."
)

print(

    "4. Scenario predictions are hypothetical "
    "and not validated future forecasts."
)

print(

    "5. Correlation results are descriptive "
    "and do not establish statistical significance."
)

print(

    "6. Tamil Nadu is included in All India totals."
)

print(

    "7. Scenario improvements do not establish "
    "economic or operational feasibility."
)

print("\n" + "=" * 75)
print("ALL ANALYSIS AND VISUALIZATION COMPLETED")
print("=" * 75)
```
