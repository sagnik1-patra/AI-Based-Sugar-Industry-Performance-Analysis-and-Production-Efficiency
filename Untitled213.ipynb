```python
# ============================================================
# PSO-BASED SUGAR INDUSTRY PERFORMANCE ANALYSIS
# AND PRODUCTION EFFICIENCY OPTIMIZATION
#
# Algorithm: Particle Swarm Optimization
# All generated files use the pso_ prefix
# ============================================================

import json
import yaml
import joblib
import random
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["font.size"] = 11

PREFIX = "pso_"

N_SIMULATED_SAMPLES = 6000

SWARM_SIZE = 60
ITERATIONS = 200

W_MAX = 0.90
W_MIN = 0.40

C1 = 1.80
C2 = 1.80

VELOCITY_LIMIT = 0.20
REGULARIZATION = 1e-8

# ============================================================
# 2. FILE PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Sugar Industry Performance Analysis and Production Efficiency"
)

DATASET_PATH = BASE_DIR / (
    "performance_of_sugar_industry_sugar_season_2019.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

H5_PATH = BASE_DIR / "pso_sugar_production_model.h5"
PKL_PATH = BASE_DIR / "pso_sugar_production_model.pkl"
YAML_PATH = BASE_DIR / "pso_config.yaml"
JSON_PATH = BASE_DIR / "pso_metrics.json"

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 80)
print("PSO-BASED SUGAR INDUSTRY PERFORMANCE ANALYSIS")
print("AND PRODUCTION EFFICIENCY OPTIMIZATION")
print("=" * 80)

# ============================================================
# 3. LOAD DATASET
# ============================================================

raw_df = pd.read_csv(DATASET_PATH)

raw_df.columns = raw_df.columns.str.strip()

raw_df["Parameters"] = (
    raw_df["Parameters"]
    .astype(str)
    .str.strip()
)

print("\nOriginal Dataset Shape:", raw_df.shape)

print("\nOriginal Dataset:")
print(raw_df.to_string(index=False))

# ============================================================
# 4. DATA TRANSFORMATION
# ============================================================

parameter_mapping = {

    "Number of Factories Working":
        "working_factories",

    "Installed Crushing Capacity (in Lakh Tonnes) per day":
        "installed_crushing_capacity",

    "Cane Crushed (in Lakh M.T.)":
        "cane_crushed",

    "Sugar Production from Cane (in Lakh M.T.)":
        "sugar_production",

    "Recovery Rate (in Percentage)":
        "recovery_rate"
}

df = raw_df.set_index("Parameters")

df = df.drop(
    columns=["S.No"],
    errors="ignore"
)

df = df.rename(index=parameter_mapping)

df = df.T.reset_index()

df = df.rename(
    columns={"index": "region_season"}
)

df.columns.name = None

required_columns = [
    "working_factories",
    "installed_crushing_capacity",
    "cane_crushed",
    "sugar_production",
    "recovery_rate"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# ============================================================
# 5. EXTRACT REGION AND SEASON
# ============================================================

df["region"] = np.where(

    df["region_season"].str.contains(
        "Tamil Nadu",
        case=False
    ),

    "Tamil Nadu",

    "All India"
)

df["season"] = df["region_season"].str.extract(
    r"(\d{4}-\d{2})"
)

df["year"] = pd.to_numeric(

    df["season"].str[:4],

    errors="coerce"
)

df["provisional"] = (
    df["region_season"]
    .str.contains(
        "Provisional",
        case=False
    )
)

for column in required_columns:

    df[column] = pd.to_numeric(

        df[column]
        .astype(str)
        .str.replace(",", "", regex=False),

        errors="coerce"
    )

df = df.dropna(

    subset=[
        "working_factories",
        "cane_crushed",
        "sugar_production",
        "recovery_rate",
        "year"
    ]

).reset_index(drop=True)

if df.empty:
    raise ValueError("No valid observations remain.")

df["label"] = (
    df["region"] +
    "\n" +
    df["season"]
)

print("\nProcessed Dataset Shape:", df.shape)

# ============================================================
# 6. FEATURE ENGINEERING
# ============================================================

df["production_per_factory"] = (

    df["sugar_production"] /
    df["working_factories"]
)

df["cane_per_factory"] = (

    df["cane_crushed"] /
    df["working_factories"]
)

df["effective_recovery_rate"] = (

    df["sugar_production"] /
    df["cane_crushed"]

) * 100

df["recovery_gap"] = (

    df["effective_recovery_rate"] -
    df["recovery_rate"]
)

df["mathematical_prediction"] = (

    df["cane_crushed"] *
    df["recovery_rate"] / 100
)

print("\nProcessed Dataset:")
print(df.to_string(index=False))

# ============================================================
# 7. GENERATE SIMULATED TRAINING DATA
# ============================================================

# The original dataset contains only four observations.
#
# Generate hypothetical samples using:
#
# Sugar Production = Cane Crushed * Recovery Rate / 100
#
# These are simulated records, not additional real data.
#
# The same simulation procedure and seed are used
# as in the previous AIS implementation.

simulation_records = []

for i in range(N_SIMULATED_SAMPLES):

    reference = df.iloc[
        rng.integers(0, len(df))
    ]

    factories = max(

        1,

        int(
            round(
                reference["working_factories"] *
                rng.uniform(0.80, 1.20)
            )
        )
    )

    cane_per_factory = (

        reference["cane_crushed"] /
        reference["working_factories"]
    )

    cane_crushed = (

        factories *
        cane_per_factory *
        rng.uniform(0.70, 1.30)
    )

    recovery_rate = np.clip(

        reference["recovery_rate"] *
        rng.uniform(0.85, 1.15),

        6.0,

        13.0
    )

    sugar_production = (

        cane_crushed *
        recovery_rate / 100
    )

    simulation_records.append({

        "working_factories": factories,

        "cane_crushed": cane_crushed,

        "recovery_rate": recovery_rate,

        "sugar_production": sugar_production
    })

simulated_df = pd.DataFrame(
    simulation_records
)

print(
    "\nSimulated Dataset Shape:",
    simulated_df.shape
)

# ============================================================
# 8. FEATURE AND TARGET SELECTION
# ============================================================

FEATURES = [

    "working_factories",

    "cane_crushed",

    "recovery_rate"
]

TARGET = "sugar_production"

X = simulated_df[FEATURES].to_numpy(
    dtype=np.float64
)

y = simulated_df[TARGET].to_numpy(
    dtype=np.float64
)

X_real = df[FEATURES].to_numpy(
    dtype=np.float64
)

y_real = df[TARGET].to_numpy(
    dtype=np.float64
)

# ============================================================
# 9. TRAIN, VALIDATION AND TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(

        X_train_full,

        y_train_full,

        test_size=0.20,

        random_state=SEED
    )
)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))
print("Original Observations:", len(X_real))

# ============================================================
# 10. POLYNOMIAL FEATURE TRANSFORMATION
# ============================================================

POLYNOMIAL_FEATURES = [

    "working_factories",

    "cane_crushed",

    "recovery_rate",

    "factories_squared",

    "cane_squared",

    "recovery_squared",

    "factories_x_cane",

    "factories_x_recovery",

    "cane_x_recovery"
]

def polynomial_transform(data):

    data = np.asarray(
        data,
        dtype=np.float64
    )

    data = np.atleast_2d(data)

    f = data[:, 0]
    c = data[:, 1]
    r = data[:, 2]

    return np.column_stack([

        f,

        c,

        r,

        f ** 2,

        c ** 2,

        r ** 2,

        f * c,

        f * r,

        c * r
    ])

X_train_poly = polynomial_transform(X_train)
X_val_poly = polynomial_transform(X_val)
X_test_poly = polynomial_transform(X_test)
X_real_poly = polynomial_transform(X_real)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_poly
)

X_val_scaled = scaler.transform(
    X_val_poly
)

X_test_scaled = scaler.transform(
    X_test_poly
)

X_real_scaled = scaler.transform(
    X_real_poly
)

y_mean = float(np.mean(y_train))
y_std = float(np.std(y_train))

if y_std == 0:
    y_std = 1.0

y_train_scaled = (
    y_train - y_mean
) / y_std

y_val_scaled = (
    y_val - y_mean
) / y_std

def add_bias(data):

    return np.column_stack([

        data,

        np.ones(len(data))
    ])

X_train_bias = add_bias(X_train_scaled)
X_val_bias = add_bias(X_val_scaled)
X_test_bias = add_bias(X_test_scaled)
X_real_bias = add_bias(X_real_scaled)

N_PARAMETERS = X_train_bias.shape[1]

print("\nPSO Particle Dimensions:", N_PARAMETERS)

# ============================================================
# 11. PARTICLE SWARM OPTIMIZATION ALGORITHM
# ============================================================

def particle_swarm_optimization(

    objective,

    lower_bounds,

    upper_bounds,

    swarm_size=60,

    iterations=200,

    w_max=0.90,

    w_min=0.40,

    c1=1.80,

    c2=1.80,

    velocity_limit=0.20,

    initial_particles=None,

    seed=42,

    verbose=True
):

    local_rng = np.random.default_rng(seed)

    lower_bounds = np.asarray(
        lower_bounds,
        dtype=float
    )

    upper_bounds = np.asarray(
        upper_bounds,
        dtype=float
    )

    dimensions = len(lower_bounds)

    search_range = (
        upper_bounds - lower_bounds
    )

    max_velocity = (
        velocity_limit * search_range
    )

    # Initialize particle positions

    positions = local_rng.uniform(

        lower_bounds,

        upper_bounds,

        size=(swarm_size, dimensions)
    )

    # Initialize particle velocities

    velocities = local_rng.uniform(

        -max_velocity,

        max_velocity,

        size=(swarm_size, dimensions)
    )

    # Optional warm-start initialization

    if initial_particles is not None:

        initial_particles = np.atleast_2d(
            initial_particles
        )

        n_initial = min(

            len(initial_particles),

            swarm_size
        )

        positions[:n_initial] = np.clip(

            initial_particles[:n_initial],

            lower_bounds,

            upper_bounds
        )

        velocities[:n_initial] = 0.0

    # Evaluate initial swarm

    fitness = np.array([

        objective(particle)

        for particle in positions
    ])

    # Personal best positions

    personal_best_positions = positions.copy()

    personal_best_fitness = fitness.copy()

    # Global best position

    best_index = np.argmin(
        personal_best_fitness
    )

    global_best_position = (
        personal_best_positions[
            best_index
        ].copy()
    )

    global_best_fitness = float(
        personal_best_fitness[best_index]
    )

    history = []

    # ========================================================
    # PSO ITERATION LOOP
    # ========================================================

    for iteration in range(iterations):

        # Adaptive inertia weight

        progress = (

            iteration /
            max(iterations - 1, 1)
        )

        inertia = (

            w_max -

            (w_max - w_min) * progress
        )

        # Random cognitive and social coefficients

        r1 = local_rng.random(
            (swarm_size, dimensions)
        )

        r2 = local_rng.random(
            (swarm_size, dimensions)
        )

        # Cognitive component

        cognitive_component = (

            c1 *

            r1 *

            (
                personal_best_positions -
                positions
            )
        )

        # Social component

        social_component = (

            c2 *

            r2 *

            (
                global_best_position -
                positions
            )
        )

        # Update velocities

        velocities = (

            inertia * velocities +

            cognitive_component +

            social_component
        )

        velocities = np.clip(

            velocities,

            -max_velocity,

            max_velocity
        )

        # Update particle positions

        new_positions = (
            positions + velocities
        )

        boundary_hits = (

            (new_positions < lower_bounds) |

            (new_positions > upper_bounds)
        )

        positions = np.clip(

            new_positions,

            lower_bounds,

            upper_bounds
        )

        velocities[boundary_hits] *= -0.5

        # Evaluate updated particles

        fitness = np.array([

            objective(particle)

            for particle in positions
        ])

        # Update personal best

        improved = (

            fitness <
            personal_best_fitness
        )

        personal_best_positions[improved] = (

            positions[improved]
        )

        personal_best_fitness[improved] = (

            fitness[improved]
        )

        # Update global best

        best_index = np.argmin(
            personal_best_fitness
        )

        if (

            personal_best_fitness[best_index] <
            global_best_fitness
        ):

            global_best_fitness = float(

                personal_best_fitness[
                    best_index
                ]
            )

            global_best_position = (

                personal_best_positions[
                    best_index
                ].copy()
            )

        history.append(
            global_best_fitness
        )

        if verbose and (

            iteration == 0 or

            (iteration + 1) % 20 == 0 or

            iteration == iterations - 1
        ):

            print(

                f"Iteration "
                f"{iteration + 1:03d}/"
                f"{iterations} | "

                f"Inertia: {inertia:.4f} | "

                f"Best Fitness: "
                f"{global_best_fitness:.10f}"
            )

    return {

        "best_position": global_best_position,

        "best_fitness": global_best_fitness,

        "history": history,

        "objective_evaluations": int(
            swarm_size * (iterations + 1)
        )
    }

# ============================================================
# 12. PSO REGRESSION FITNESS FUNCTION
# ============================================================

# Each particle represents nine polynomial coefficients
# and one bias parameter.
#
# Use precomputed matrices for faster fitness evaluation.

train_gram = (

    X_train_bias.T @ X_train_bias

) / len(X_train_bias)

train_cross = (

    X_train_bias.T @ y_train_scaled

) / len(X_train_bias)

train_constant = np.mean(
    y_train_scaled ** 2
)

val_gram = (

    X_val_bias.T @ X_val_bias

) / len(X_val_bias)

val_cross = (

    X_val_bias.T @ y_val_scaled

) / len(X_val_bias)

val_constant = np.mean(
    y_val_scaled ** 2
)

def model_fitness(particle):

    train_mse = max(

        0.0,

        particle @ train_gram @ particle -

        2 * particle @ train_cross +

        train_constant
    )

    val_mse = max(

        0.0,

        particle @ val_gram @ particle -

        2 * particle @ val_cross +

        val_constant
    )

    regularization_penalty = (

        REGULARIZATION *

        np.sum(particle[:-1] ** 2)
    )

    fitness = (

        0.70 * train_mse +

        0.30 * val_mse +

        regularization_penalty
    )

    return float(fitness)

# ============================================================
# 13. TRAIN PSO MODEL
# ============================================================

print("\n" + "=" * 80)
print("TRAINING PSO REGRESSION MODEL")
print("=" * 80)

lower_bounds = np.full(
    N_PARAMETERS,
    -5.0
)

upper_bounds = np.full(
    N_PARAMETERS,
    5.0
)

# Least-squares warm start, matching the AIS approach.

least_squares_weights = np.linalg.lstsq(

    X_train_bias,

    y_train_scaled,

    rcond=None

)[0]

pso_training = particle_swarm_optimization(

    objective=model_fitness,

    lower_bounds=lower_bounds,

    upper_bounds=upper_bounds,

    swarm_size=SWARM_SIZE,

    iterations=ITERATIONS,

    w_max=W_MAX,

    w_min=W_MIN,

    c1=C1,

    c2=C2,

    velocity_limit=VELOCITY_LIMIT,

    initial_particles=[
        least_squares_weights
    ],

    seed=SEED,

    verbose=True
)

best_weights = pso_training[
    "best_position"
]

training_history = pso_training[
    "history"
]

print("\nBest PSO Fitness:")

print(
    pso_training["best_fitness"]
)

print("\nOptimized PSO Weights:")

print(best_weights)

# ============================================================
# 14. PSO PREDICTION FUNCTION
# ============================================================

def pso_predict(data):

    data = np.asarray(
        data,
        dtype=np.float64
    )

    polynomial_data = (
        polynomial_transform(data)
    )

    scaled_data = scaler.transform(
        polynomial_data
    )

    bias_data = add_bias(
        scaled_data
    )

    normalized_prediction = (

        bias_data @ best_weights
    )

    prediction = (

        normalized_prediction *
        y_std +

        y_mean
    )

    return np.maximum(
        prediction,
        0.0
    )

# ============================================================
# 15. GENERATE MODEL PREDICTIONS
# ============================================================

pso_train_predictions = pso_predict(
    X_train
)

pso_val_predictions = pso_predict(
    X_val
)

pso_test_predictions = pso_predict(
    X_test
)

pso_real_predictions = pso_predict(
    X_real
)

baseline_test_predictions = (

    X_test[:, 1] *
    X_test[:, 2] / 100
)

baseline_real_predictions = (

    X_real[:, 1] *
    X_real[:, 2] / 100
)

df["pso_prediction"] = (
    pso_real_predictions
)

df["pso_absolute_error"] = np.abs(

    df["sugar_production"] -

    df["pso_prediction"]
)

# ============================================================
# 16. MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(
        actual,
        dtype=float
    ).flatten()

    predicted = np.asarray(
        predicted,
        dtype=float
    ).flatten()

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    nonzero = actual != 0

    if np.any(nonzero):

        mape = np.mean(

            np.abs(

                (
                    actual[nonzero] -

                    predicted[nonzero]
                ) /

                actual[nonzero]
            )

        ) * 100

        estimation_score = 100 - mape

    else:

        mape = None
        estimation_score = None

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2_Score": float(r2),

        "MAPE_Percentage": (

            float(mape)

            if mape is not None

            else None
        ),

        "Estimation_Score_Percentage": (

            float(estimation_score)

            if estimation_score is not None

            else None
        )
    }

# ============================================================
# 17. PSO PERFORMANCE EVALUATION
# ============================================================

pso_metrics = {

    "training": evaluate_model(

        y_train,

        pso_train_predictions
    ),

    "validation": evaluate_model(

        y_val,

        pso_val_predictions
    ),

    "simulation_test": evaluate_model(

        y_test,

        pso_test_predictions
    ),

    "original_observations": evaluate_model(

        y_real,

        pso_real_predictions
    )
}

baseline_metrics = {

    "simulation_test": evaluate_model(

        y_test,

        baseline_test_predictions
    ),

    "original_observations": evaluate_model(

        y_real,

        baseline_real_predictions
    )
}

print("\n" + "=" * 80)
print("PSO MODEL PERFORMANCE")
print("=" * 80)

for dataset_name, values in pso_metrics.items():

    print(f"\n{dataset_name.upper()}")

    for metric_name, value in values.items():

        if value is not None:

            print(
                f"{metric_name}: {value:.8f}"
            )

# ============================================================
# 18. LOAD PREVIOUS MODELS FOR COMPARISON
# ============================================================

comparison_results = {

    "Mathematical Baseline":
        baseline_metrics["original_observations"]
}

simulation_comparison_results = {

    "Mathematical Baseline":
        baseline_metrics["simulation_test"]
}

rf_model = None
ann_model = None
ais_package = None

# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------

previous_rf_path = (

    BASE_DIR /
    "sugar_production_model.pkl"
)

if previous_rf_path.exists():

    try:

        rf_model = joblib.load(
            previous_rf_path
        )

        rf_real_predictions = rf_model.predict(

            pd.DataFrame(
                X_real,
                columns=FEATURES
            )
        )

        rf_test_predictions = rf_model.predict(

            pd.DataFrame(
                X_test,
                columns=FEATURES
            )
        )

        comparison_results["Random Forest"] = (
            evaluate_model(
                y_real,
                rf_real_predictions
            )
        )

        simulation_comparison_results[
            "Random Forest"
        ] = evaluate_model(

            y_test,

            rf_test_predictions
        )

        df["random_forest_prediction"] = (
            rf_real_predictions
        )

        print("\nRandom Forest model loaded.")

    except Exception as error:

        print("RF loading error:", error)

        rf_model = None

# ------------------------------------------------------------
# ANN
# ------------------------------------------------------------

previous_ann_path = (

    BASE_DIR /
    "sugar_production_model.h5"
)

if previous_ann_path.exists():

    try:

        ann_model = tf.keras.models.load_model(

            str(previous_ann_path),

            compile=False
        )

        ann_real_predictions = ann_model.predict(

            X_real.astype(np.float32),

            verbose=0

        ).flatten()

        ann_test_predictions = ann_model.predict(

            X_test.astype(np.float32),

            verbose=0

        ).flatten()

        comparison_results["ANN"] = evaluate_model(

            y_real,

            ann_real_predictions
        )

        simulation_comparison_results["ANN"] = (
            evaluate_model(

                y_test,

                ann_test_predictions
            )
        )

        df["ann_prediction"] = (
            ann_real_predictions
        )

        print("\nANN model loaded.")

    except Exception as error:

        print("ANN loading error:", error)

        ann_model = None

# ------------------------------------------------------------
# AIS
# ------------------------------------------------------------

previous_ais_path = (

    BASE_DIR /
    "ais_sugar_production_model.pkl"
)

def predict_saved_ais(data, package):

    polynomial_data = polynomial_transform(
        data
    )

    scaled_data = package[
        "scaler"
    ].transform(polynomial_data)

    bias_data = add_bias(
        scaled_data
    )

    prediction = (

        (
            bias_data @ package["weights"]
        ) *

        package["target_std"] +

        package["target_mean"]
    )

    return np.maximum(
        prediction,
        0.0
    )

if previous_ais_path.exists():

    try:

        ais_package = joblib.load(
            previous_ais_path
        )

        ais_real_predictions = predict_saved_ais(

            X_real,

            ais_package
        )

        ais_test_predictions = predict_saved_ais(

            X_test,

            ais_package
        )

        comparison_results["AIS"] = evaluate_model(

            y_real,

            ais_real_predictions
        )

        simulation_comparison_results["AIS"] = (
            evaluate_model(

                y_test,

                ais_test_predictions
            )
        )

        df["ais_prediction"] = (
            ais_real_predictions
        )

        print("\nAIS model loaded.")

    except Exception as error:

        print("AIS loading error:", error)

        ais_package = None

# Add PSO results

comparison_results["PSO"] = (

    pso_metrics["original_observations"]
)

simulation_comparison_results["PSO"] = (

    pso_metrics["simulation_test"]
)

comparison_df = pd.DataFrame(
    comparison_results
).T

comparison_df.index.name = "Model"

simulation_comparison_df = pd.DataFrame(
    simulation_comparison_results
).T

simulation_comparison_df.index.name = "Model"

comparison_df.to_csv(

    BASE_DIR / "pso_model_comparison.csv"
)

simulation_comparison_df.to_csv(

    BASE_DIR / "pso_simulation_comparison.csv"
)

print("\nOriginal Observation Comparison:")

print(comparison_df.round(6))

print("\nSimulation Test Comparison:")

print(simulation_comparison_df.round(6))

# ============================================================
# 19. SAVE PSO MODEL AS PKL
# ============================================================

pso_model_package = {

    "algorithm": "Particle Swarm Optimization",

    "features": FEATURES,

    "polynomial_features": POLYNOMIAL_FEATURES,

    "scaler": scaler,

    "weights": best_weights,

    "target_mean": y_mean,

    "target_std": y_std,

    "training_data_type": "Simulated",

    "random_seed": SEED
}

joblib.dump(

    pso_model_package,

    PKL_PATH
)

print("\n[SAVED]", PKL_PATH.name)

# ============================================================
# 20. SAVE PSO MODEL AS KERAS H5
# ============================================================

# Convert the PSO polynomial regression equation
# into an equivalent Keras model.
#
# No gradient-based training is performed.

inputs = tf.keras.Input(

    shape=(3,),

    name="raw_industry_features"
)

reshaped = tf.keras.layers.Reshape(
    (3, 1)
)(inputs)

f = tf.keras.layers.Flatten()(

    tf.keras.layers.Cropping1D(
        cropping=(0, 2)
    )(reshaped)
)

c = tf.keras.layers.Flatten()(

    tf.keras.layers.Cropping1D(
        cropping=(1, 1)
    )(reshaped)
)

r = tf.keras.layers.Flatten()(

    tf.keras.layers.Cropping1D(
        cropping=(2, 0)
    )(reshaped)
)

f_squared = tf.keras.layers.Multiply()(
    [f, f]
)

c_squared = tf.keras.layers.Multiply()(
    [c, c]
)

r_squared = tf.keras.layers.Multiply()(
    [r, r]
)

fc = tf.keras.layers.Multiply()(
    [f, c]
)

fr = tf.keras.layers.Multiply()(
    [f, r]
)

cr = tf.keras.layers.Multiply()(
    [c, r]
)

polynomial_layer = tf.keras.layers.Concatenate(

    name="polynomial_features"

)([

    f,

    c,

    r,

    f_squared,

    c_squared,

    r_squared,

    fc,

    fr,

    cr
])

# Fold normalization into the Dense layer.

raw_coefficients = (

    best_weights[:-1] *

    y_std /

    scaler.scale_
)

raw_bias = (

    best_weights[-1] * y_std +

    y_mean -

    np.sum(

        raw_coefficients *
        scaler.mean_
    )
)

output_layer = tf.keras.layers.Dense(

    1,

    activation="relu",

    name="pso_optimized_regression"
)

outputs = output_layer(
    polynomial_layer
)

pso_keras_model = tf.keras.Model(

    inputs=inputs,

    outputs=outputs,

    name="PSO_Sugar_Production_Model"
)

output_layer.set_weights([

    raw_coefficients.reshape(
        -1, 1
    ).astype(np.float32),

    np.array(
        [raw_bias],
        dtype=np.float32
    )
])

pso_keras_model.save(

    str(H5_PATH),

    include_optimizer=False
)

print("\n[SAVED]", H5_PATH.name)

# Verify H5 model

loaded_h5_model = tf.keras.models.load_model(

    str(H5_PATH),

    compile=False
)

h5_predictions = loaded_h5_model.predict(

    X_real.astype(np.float32),

    verbose=0

).flatten()

maximum_difference = float(

    np.max(

        np.abs(

            h5_predictions -

            pso_real_predictions
        )
    )
)

print(

    "\nMaximum H5 verification difference:",

    maximum_difference
)

# ============================================================
# 21. GRAPH SAVING FUNCTION
# ============================================================

def save_and_show(filename):

    if not filename.startswith(PREFIX):

        filename = PREFIX + filename

    plt.tight_layout()

    plt.savefig(

        BASE_DIR / filename,

        dpi=300,

        bbox_inches="tight",

        facecolor="white"
    )

    plt.show()

    plt.close()

    print(f"[SAVED] {filename}")

def annotate_bars(ax, bars, decimals=4, suffix=""):

    for bar in bars:

        height = bar.get_height()

        ax.annotate(

            f"{height:.{decimals}f}{suffix}",

            (
                bar.get_x() +
                bar.get_width() / 2,

                height
            ),

            xytext=(0, 5),

            textcoords="offset points",

            ha="center",

            fontsize=9
        )

# ============================================================
# 22. PSO ACCURACY GRAPH
# ============================================================

# Regression estimation score = 100 - MAPE.
# This is not classification accuracy.

dataset_names = [

    "Training",

    "Validation",

    "Simulation Test",

    "Original Observations"
]

accuracy_values = [

    pso_metrics["training"][
        "Estimation_Score_Percentage"
    ],

    pso_metrics["validation"][
        "Estimation_Score_Percentage"
    ],

    pso_metrics["simulation_test"][
        "Estimation_Score_Percentage"
    ],

    pso_metrics["original_observations"][
        "Estimation_Score_Percentage"
    ]
]

fig, ax = plt.subplots(
    figsize=(12, 7)
)

bars = ax.bar(

    dataset_names,

    accuracy_values,

    color=[
        "#3498db",
        "#2ecc71",
        "#9b59b6",
        "#e67e22"
    ],

    edgecolor="black"
)

annotate_bars(
    ax,
    bars,
    decimals=4,
    suffix="%"
)

ax.set_title(

    "PSO-Based Sugar Production Estimation Performance\n"
    "(100 - MAPE; Not Classification Accuracy)",

    fontsize=16,

    fontweight="bold"
)

ax.set_ylabel(
    "Estimation Score (%)"
)

ax.set_ylim(

    min(0, min(accuracy_values) - 10),

    max(110, max(accuracy_values) + 10)
)

ax.grid(
    axis="y",
    alpha=0.3
)

save_and_show(
    "pso_accuracy_graph.png"
)

# ============================================================
# 23. PSO CORRELATION HEATMAP
# ============================================================

heatmap_columns = [

    "working_factories",

    "installed_crushing_capacity",

    "cane_crushed",

    "sugar_production",

    "recovery_rate",

    "production_per_factory",

    "cane_per_factory",

    "effective_recovery_rate",

    "pso_prediction"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(
    figsize=(14, 11)
)

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".2f",

    cmap="coolwarm",

    center=0,

    vmin=-1,

    vmax=1,

    linewidths=0.5,

    square=True,

    cbar_kws={

        "label": "Correlation Coefficient"
    }
)

plt.title(

    "PSO Sugar Industry Feature Correlation Heatmap\n"
    "Descriptive Analysis of Original Observations",

    fontsize=16,

    fontweight="bold"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.yticks(rotation=0)

save_and_show(
    "pso_heatmap.png"
)

# ============================================================
# 24. PSO MODEL COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(

    2,

    2,

    figsize=(17, 12)
)

comparison_metrics = [

    "MAE",

    "RMSE",

    "R2_Score",

    "MAPE_Percentage"
]

metric_titles = [

    "Mean Absolute Error",

    "Root Mean Squared Error",

    "R² Score",

    "Mean Absolute Percentage Error"
]

model_colors = {

    "Mathematical Baseline": "#34495e",

    "Random Forest": "#3498db",

    "ANN": "#2ecc71",

    "AIS": "#9b59b6",

    "PSO": "#e74c3c"
}

for ax, metric, title in zip(

    axes.flatten(),

    comparison_metrics,

    metric_titles
):

    colors = [

        model_colors.get(
            model,
            "#3498db"
        )

        for model in comparison_df.index
    ]

    bars = ax.bar(

        comparison_df.index,

        comparison_df[metric],

        color=colors,

        edgecolor="black"
    )

    annotate_bars(
        ax,
        bars
    )

    ax.set_title(

        title,

        fontweight="bold"
    )

    ax.tick_params(

        axis="x",

        rotation=20
    )

    ax.margins(y=0.20)

    ax.grid(

        axis="y",

        alpha=0.3
    )

fig.suptitle(

    "Comparative Performance Analysis Including PSO\n"
    "Original Region-Season Observations",

    fontsize=18,

    fontweight="bold"
)

save_and_show(
    "pso_comparison_graph.png"
)

# ============================================================
# 25. PSO CONVERGENCE GRAPH
# ============================================================

plt.figure(
    figsize=(12, 7)
)

plt.plot(

    np.arange(
        1,
        len(training_history) + 1
    ),

    np.maximum(
        training_history,
        1e-16
    ),

    color="#e74c3c",

    linewidth=2.5,

    label="PSO Best Fitness"
)

plt.yscale("log")

plt.xlabel("Iteration")

plt.ylabel(
    "Regularized Prediction Error"
)

plt.title(

    "Particle Swarm Optimization Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.legend()

plt.grid(alpha=0.3)

save_and_show(
    "pso_convergence_graph.png"
)

# ============================================================
# 26. PSO PRODUCTION OPTIMIZATION
# ============================================================

# Decision variables:
#
# x[0] = relative change in working factories
# x[1] = relative change in cane crushed
# x[2] = relative change in recovery rate
#
# Use the same hypothetical constraints as AIS.

MAX_FACTORY_CHANGE = 0.05
MAX_CANE_CHANGE = 0.10
MAX_RECOVERY_CHANGE = 0.05

MAX_RECOVERY_RATE = 13.0

INTERVENTION_BUDGET = 1.50

OPTIMIZATION_ITERATIONS = 150

optimization_records = []

optimization_histories = {}

optimization_evaluations = {}

def intervention_cost(particle):

    return float(

        particle[0] / MAX_FACTORY_CHANGE +

        particle[1] / MAX_CANE_CHANGE +

        particle[2] / MAX_RECOVERY_CHANGE
    )

def repair_budget(particle):

    particle = np.asarray(

        particle,

        dtype=float
    ).copy()

    particle = np.clip(

        particle,

        [0.0, 0.0, 0.0],

        [
            MAX_FACTORY_CHANGE,
            MAX_CANE_CHANGE,
            MAX_RECOVERY_CHANGE
        ]
    )

    cost = intervention_cost(
        particle
    )

    if cost > INTERVENTION_BUDGET:

        particle *= (

            INTERVENTION_BUDGET /
            cost
        )

    return particle

def decode_solution(reference, particle):

    factory_change = float(
        particle[0]
    )

    cane_change = float(
        particle[1]
    )

    recovery_change = float(
        particle[2]
    )

    factories = max(

        1,

        int(

            np.floor(

                reference["working_factories"] *

                (1 + factory_change) +

                1e-9
            )
        )
    )

    cane_crushed = (

        reference["cane_crushed"] *

        (1 + cane_change)
    )

    recovery_rate = min(

        reference["recovery_rate"] *

        (1 + recovery_change),

        MAX_RECOVERY_RATE
    )

    return (

        factories,

        float(cane_crushed),

        float(recovery_rate)
    )

for index, reference in df.iterrows():

    region = reference["region"]

    season = reference["season"]

    baseline_production = (

        reference["cane_crushed"] *

        reference["recovery_rate"] / 100
    )

    baseline_cane_per_factory = (

        reference["cane_crushed"] /

        reference["working_factories"]
    )

    def production_fitness(particle):

        particle = repair_budget(
            particle
        )

        factories, cane, recovery = (
            decode_solution(

                reference,

                particle
            )
        )

        production = (

            cane *
            recovery / 100
        )

        maximum_cane_per_factory = (

            baseline_cane_per_factory *

            1.10
        )

        capacity_violation = max(

            0.0,

            (

                cane / factories -

                maximum_cane_per_factory

            ) /

            maximum_cane_per_factory
        )

        production_gain = (

            production /
            baseline_production - 1
        )

        objective = (

            -production_gain +

            0.005 * intervention_cost(
                particle
            ) +

            1000.0 * capacity_violation
        )

        return float(objective)

    print("\n" + "-" * 80)

    print(
        f"PSO OPTIMIZING: {region} | {season}"
    )

    optimization = particle_swarm_optimization(

        objective=production_fitness,

        lower_bounds=[
            0.0,
            0.0,
            0.0
        ],

        upper_bounds=[

            MAX_FACTORY_CHANGE,

            MAX_CANE_CHANGE,

            MAX_RECOVERY_CHANGE
        ],

        swarm_size=50,

        iterations=OPTIMIZATION_ITERATIONS,

        w_max=W_MAX,

        w_min=W_MIN,

        c1=C1,

        c2=C2,

        velocity_limit=VELOCITY_LIMIT,

        initial_particles=[
            [0.0, 0.0, 0.0]
        ],

        seed=SEED + index,

        verbose=False
    )

    best_solution = repair_budget(

        optimization["best_position"]
    )

    optimized_factories, optimized_cane, optimized_recovery = (
        decode_solution(

            reference,

            best_solution
        )
    )

    # Repair any residual processing constraint violation.

    maximum_allowed_cane = (

        optimized_factories *

        baseline_cane_per_factory *

        1.10
    )

    optimized_cane = min(

        optimized_cane,

        maximum_allowed_cane
    )

    best_solution[1] = max(

        0.0,

        optimized_cane /

        reference["cane_crushed"] - 1
    )

    optimized_production = (

        optimized_cane *

        optimized_recovery / 100
    )

    optimized_pso_prediction = float(

        pso_predict([[

            optimized_factories,

            optimized_cane,

            optimized_recovery

        ]])[0]
    )

    additional_production = (

        optimized_production -

        baseline_production
    )

    improvement_percentage = (

        additional_production /

        baseline_production

    ) * 100

    budget_used = intervention_cost(
        best_solution
    )

    optimization_records.append({

        "region": region,

        "season": season,

        "provisional": bool(
            reference["provisional"]
        ),

        "original_factories": int(
            reference["working_factories"]
        ),

        "pso_optimized_factories": int(
            optimized_factories
        ),

        "original_cane_crushed": float(
            reference["cane_crushed"]
        ),

        "pso_optimized_cane_crushed": float(
            optimized_cane
        ),

        "original_recovery_rate": float(
            reference["recovery_rate"]
        ),

        "pso_optimized_recovery_rate": float(
            optimized_recovery
        ),

        "reported_sugar_production": float(
            reference["sugar_production"]
        ),

        "baseline_calculated_production": float(
            baseline_production
        ),

        "pso_original_prediction": float(
            reference["pso_prediction"]
        ),

        "pso_optimized_production": float(
            optimized_production
        ),

        "pso_optimized_prediction": float(
            optimized_pso_prediction
        ),

        "additional_production": float(
            additional_production
        ),

        "improvement_percentage": float(
            improvement_percentage
        ),

        "intervention_budget_used": float(
            budget_used
        ),

        "within_budget": bool(

            budget_used <=
            INTERVENTION_BUDGET + 1e-8
        ),

        "optimization_fitness": float(

            production_fitness(
                best_solution
            )
        ),

        "objective_evaluations": int(

            optimization[
                "objective_evaluations"
            ]
        ),

        "data_type": "Hypothetical Optimization"
    })

    key = f"{region} {season}"

    optimization_histories[key] = (

        optimization["history"]
    )

    optimization_evaluations[key] = (

        optimization["objective_evaluations"]
    )

    print(

        f"Baseline Production: "
        f"{baseline_production:.4f} Lakh MT"
    )

    print(

        f"PSO Optimized Production: "
        f"{optimized_production:.4f} Lakh MT"
    )

    print(

        f"Production Improvement: "
        f"{improvement_percentage:.4f}%"
    )

# ============================================================
# 27. SAVE PSO RESULT CSV
# ============================================================

result_df = pd.DataFrame(
    optimization_records
)

result_df["original_production_per_factory"] = (

    result_df["reported_sugar_production"] /

    result_df["original_factories"]
)

result_df["pso_production_per_factory"] = (

    result_df["pso_optimized_production"] /

    result_df["pso_optimized_factories"]
)

# Load AIS production optimization results if available.

ais_result_path = (

    BASE_DIR /
    "ais_result.csv"
)

if ais_result_path.exists():

    try:

        ais_result_df = pd.read_csv(
            ais_result_path
        )

        ais_comparison = ais_result_df[[

            "region",

            "season",

            "ais_optimized_production",

            "improvement_percentage"

        ]].rename(columns={

            "improvement_percentage":
                "ais_improvement_percentage"
        })

        result_df = result_df.merge(

            ais_comparison,

            on=["region", "season"],

            how="left",

            validate="one_to_one"
        )

        result_df["pso_minus_ais_production"] = (

            result_df["pso_optimized_production"] -

            result_df["ais_optimized_production"]
        )

        print("\nAIS optimization results included.")

    except Exception as error:

        print(
            "AIS result comparison error:",
            error
        )

result_df.round(8).to_csv(

    BASE_DIR / "pso_result.csv",

    index=False
)

print("\n" + "=" * 80)
print("PSO OPTIMIZATION RESULTS")
print("=" * 80)

print(

    result_df.round(4).to_string(
        index=False
    )
)

print("\n[SAVED] pso_result.csv")

# ============================================================
# 28. PSO RESULT GRAPH
# ============================================================

regions = result_df["region"].unique()

fig, axes = plt.subplots(

    1,

    len(regions),

    figsize=(17, 8),

    squeeze=False
)

for ax, region in zip(

    axes.flatten(),

    regions
):

    regional_data = result_df[

        result_df["region"] == region

    ].sort_values("season")

    columns = [

        "reported_sugar_production",

        "baseline_calculated_production",

        "pso_original_prediction",

        "pso_optimized_production"
    ]

    labels = [

        "Reported Production",

        "Mathematical Baseline",

        "PSO Prediction",

        "PSO Optimized Production"
    ]

    colors = [

        "#34495e",

        "#3498db",

        "#e67e22",

        "#e74c3c"
    ]

    if "ais_optimized_production" in regional_data:

        columns.append(
            "ais_optimized_production"
        )

        labels.append(
            "AIS Optimized Production"
        )

        colors.append(
            "#9b59b6"
        )

    positions = np.arange(
        len(regional_data)
    )

    width = 0.80 / len(columns)

    for i, (column, label, color) in enumerate(

        zip(
            columns,
            labels,
            colors
        )
    ):

        offset = (

            i -
            (len(columns) - 1) / 2

        ) * width

        bars = ax.bar(

            positions + offset,

            regional_data[column],

            width,

            label=label,

            color=color,

            edgecolor="black"
        )

        for bar in bars:

            height = bar.get_height()

            if np.isfinite(height):

                ax.annotate(

                    f"{height:.2f}",

                    (
                        bar.get_x() +
                        bar.get_width() / 2,

                        height
                    ),

                    xytext=(0, 4),

                    textcoords="offset points",

                    ha="center",

                    rotation=90,

                    fontsize=8
                )

    ax.set_xticks(positions)

    ax.set_xticklabels(
        regional_data["season"]
    )

    ax.set_title(

        region,

        fontweight="bold"
    )

    ax.set_ylabel(

        "Sugar Production (Lakh MT)"
    )

    ax.margins(y=0.30)

    ax.legend(fontsize=8)

fig.suptitle(

    "PSO-Based Sugar Production Efficiency Optimization",

    fontsize=18,

    fontweight="bold"
)

save_and_show(
    "pso_result_graph.png"
)

# ============================================================
# 29. PSO PRODUCTION OPTIMIZATION CONVERGENCE
# ============================================================

plt.figure(
    figsize=(13, 8)
)

for label, history in optimization_histories.items():

    plt.plot(

        np.arange(
            1,
            len(history) + 1
        ),

        history,

        linewidth=2,

        label=label
    )

plt.title(

    "PSO Production Optimization Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel(

    "Optimization Objective (Lower is Better)"
)

plt.legend()

plt.grid(alpha=0.3)

save_and_show(
    "pso_optimization_graph.png"
)

# ============================================================
# 30. GENERATE PRODUCTION PREDICTION SCENARIOS
# ============================================================

latest_df = (

    df.sort_values("year")

    .groupby("region")

    .tail(1)

    .reset_index(drop=True)
)

scenarios = [

    {
        "scenario": "Current Conditions",
        "factory_change": 0.00,
        "cane_change": 0.00,
        "recovery_change": 0.00
    },

    {
        "scenario": "Recovery Improvement",
        "factory_change": 0.00,
        "cane_change": 0.00,
        "recovery_change": 0.05
    },

    {
        "scenario": "Increased Cane Availability",
        "factory_change": 0.00,
        "cane_change": 0.10,
        "recovery_change": 0.00
    },

    {
        "scenario": "Factory Expansion",
        "factory_change": 0.05,
        "cane_change": 0.05,
        "recovery_change": 0.00
    },

    {
        "scenario": "Combined Improvement",
        "factory_change": 0.05,
        "cane_change": 0.10,
        "recovery_change": 0.05
    },

    {
        "scenario": "Reduced Cane Availability",
        "factory_change": 0.00,
        "cane_change": -0.10,
        "recovery_change": 0.00
    }
]

prediction_records = []

for _, reference in latest_df.iterrows():

    baseline_production = (

        reference["cane_crushed"] *

        reference["recovery_rate"] / 100
    )

    for scenario in scenarios:

        factories = max(

            1,

            int(

                np.floor(

                    reference["working_factories"] *

                    (1 + scenario["factory_change"]) +

                    1e-9
                )
            )
        )

        cane = (

            reference["cane_crushed"] *

            (1 + scenario["cane_change"])
        )

        recovery = np.clip(

            reference["recovery_rate"] *

            (1 + scenario["recovery_change"]),

            6.0,

            MAX_RECOVERY_RATE
        )

        prediction_records.append({

            "region": reference["region"],

            "reference_season": reference["season"],

            "scenario": scenario["scenario"],

            "working_factories": factories,

            "cane_crushed": float(cane),

            "recovery_rate": float(recovery),

            "reference_reported_production": float(
                reference["sugar_production"]
            ),

            "reference_calculated_production": float(
                baseline_production
            ),

            "data_type": "Hypothetical Scenario"
        })

    # Add PSO optimized scenario.

    optimized_row = result_df[

        (
            result_df["region"] ==
            reference["region"]
        )

        &

        (
            result_df["season"] ==
            reference["season"]
        )

    ].iloc[0]

    prediction_records.append({

        "region": reference["region"],

        "reference_season": reference["season"],

        "scenario": "PSO Optimized Scenario",

        "working_factories": int(

            optimized_row[
                "pso_optimized_factories"
            ]
        ),

        "cane_crushed": float(

            optimized_row[
                "pso_optimized_cane_crushed"
            ]
        ),

        "recovery_rate": float(

            optimized_row[
                "pso_optimized_recovery_rate"
            ]
        ),

        "reference_reported_production": float(

            reference["sugar_production"]
        ),

        "reference_calculated_production": float(

            baseline_production
        ),

        "data_type": "PSO Hypothetical Optimization"
    })

prediction_df = pd.DataFrame(
    prediction_records
)

# ============================================================
# 31. GENERATE PSO PREDICTIONS
# ============================================================

X_future = prediction_df[
    FEATURES
].to_numpy(dtype=float)

prediction_df["mathematical_prediction"] = (

    prediction_df["cane_crushed"] *

    prediction_df["recovery_rate"] / 100
)

prediction_df["pso_prediction"] = (

    pso_predict(X_future)
)

if rf_model is not None:

    prediction_df["random_forest_prediction"] = (

        rf_model.predict(

            prediction_df[FEATURES]
        )
    )

if ann_model is not None:

    prediction_df["ann_prediction"] = (

        ann_model.predict(

            X_future.astype(np.float32),

            verbose=0

        ).flatten()
    )

if ais_package is not None:

    prediction_df["ais_prediction"] = (

        predict_saved_ais(

            X_future,

            ais_package
        )
    )

prediction_df["pso_expected_change"] = (

    prediction_df["pso_prediction"] -

    prediction_df["reference_calculated_production"]
)

prediction_df["pso_improvement_percentage"] = (

    prediction_df["pso_expected_change"] /

    prediction_df["reference_calculated_production"]

) * 100

prediction_df["formula_improvement_percentage"] = (

    (

        prediction_df["mathematical_prediction"] -

        prediction_df["reference_calculated_production"]
    ) /

    prediction_df["reference_calculated_production"]

) * 100

prediction_df["performance_category"] = np.select(

    [

        prediction_df[
            "formula_improvement_percentage"
        ] >= 10,

        prediction_df[
            "formula_improvement_percentage"
        ] > 1e-8,

        prediction_df[
            "formula_improvement_percentage"
        ] < -1e-8
    ],

    [

        "High Improvement",

        "Moderate Improvement",

        "Production Decline"
    ],

    default="Stable"
)

# ============================================================
# 32. CHECK SCENARIO CONSTRAINTS
# ============================================================

scenario_budget_values = []

scenario_capacity_checks = []

for _, row in prediction_df.iterrows():

    reference = latest_df[

        latest_df["region"] ==
        row["region"]

    ].iloc[0]

    cost = (

        max(

            0,

            row["working_factories"] /

            reference["working_factories"] - 1

        ) / MAX_FACTORY_CHANGE

        +

        max(

            0,

            row["cane_crushed"] /

            reference["cane_crushed"] - 1

        ) / MAX_CANE_CHANGE

        +

        max(

            0,

            row["recovery_rate"] /

            reference["recovery_rate"] - 1

        ) / MAX_RECOVERY_CHANGE
    )

    maximum_cane_per_factory = (

        reference["cane_crushed"] /

        reference["working_factories"]

    ) * 1.10

    within_capacity = (

        row["cane_crushed"] /

        row["working_factories"]

        <= maximum_cane_per_factory + 1e-8
    )

    scenario_budget_values.append(
        float(cost)
    )

    scenario_capacity_checks.append(
        bool(within_capacity)
    )

prediction_df["intervention_budget_used"] = (

    scenario_budget_values
)

prediction_df["within_pso_budget"] = (

    prediction_df["intervention_budget_used"]

    <= INTERVENTION_BUDGET + 1e-8
)

prediction_df["within_processing_capacity"] = (

    scenario_capacity_checks
)

prediction_df["scenario_feasible"] = (

    prediction_df["within_pso_budget"]

    &

    prediction_df["within_processing_capacity"]
)

# ============================================================
# 33. SAVE PSO PREDICTION CSV
# ============================================================

prediction_df.round(8).to_csv(

    BASE_DIR / "pso_prediction.csv",

    index=False
)

print("\n" + "=" * 80)
print("PSO SUGAR PRODUCTION PREDICTIONS")
print("=" * 80)

print(

    prediction_df.round(4).to_string(
        index=False
    )
)

print("\n[SAVED] pso_prediction.csv")

# ============================================================
# 34. PSO PREDICTION GRAPH
# ============================================================

regions = prediction_df["region"].unique()

fig, axes = plt.subplots(

    len(regions),

    1,

    figsize=(17, 7 * len(regions)),

    squeeze=False
)

prediction_models = {

    "mathematical_prediction":
        ("Mathematical Baseline", "o", "#34495e"),

    "random_forest_prediction":
        ("Random Forest", "s", "#3498db"),

    "ann_prediction":
        ("ANN", "^", "#2ecc71"),

    "ais_prediction":
        ("AIS", "D", "#9b59b6"),

    "pso_prediction":
        ("PSO", "P", "#e74c3c")
}

for ax, region in zip(

    axes.flatten(),

    regions
):

    regional_predictions = prediction_df[

        prediction_df["region"] == region

    ].reset_index(drop=True)

    positions = np.arange(

        len(regional_predictions)
    )

    for column, settings in prediction_models.items():

        if column not in regional_predictions:

            continue

        label, marker, color = settings

        ax.plot(

            positions,

            regional_predictions[column],

            marker=marker,

            linewidth=3 if label == "PSO" else 2,

            color=color,

            label=label
        )

    ax.axhline(

        regional_predictions[
            "reference_reported_production"
        ].iloc[0],

        color="black",

        linestyle="--",

        label="Reported Reference Production"
    )

    ax.set_xticks(positions)

    ax.set_xticklabels(

        regional_predictions["scenario"],

        rotation=20,

        ha="right"
    )

    ax.set_title(

        f"{region} - PSO Production Predictions",

        fontweight="bold"
    )

    ax.set_ylabel(

        "Estimated Production (Lakh MT)"
    )

    ax.legend()

    ax.grid(alpha=0.3)

fig.suptitle(

    "PSO-Based Sugar Production Scenario Analysis",

    fontsize=18,

    fontweight="bold"
)

save_and_show(
    "pso_prediction_graph.png"
)

# ============================================================
# 35. PSO FEATURE SENSITIVITY ANALYSIS
# ============================================================

base_predictions = pso_predict(
    X_test
)

feature_sensitivity = {}

for index, feature in enumerate(FEATURES):

    modified_data = X_test.copy()

    # Increase one feature by 1%.

    modified_data[:, index] *= 1.01

    modified_predictions = pso_predict(

        modified_data
    )

    relative_change = np.mean(

        np.abs(

            modified_predictions -

            base_predictions
        ) /

        np.maximum(

            np.abs(base_predictions),

            1e-8
        )

    ) * 100

    feature_sensitivity[feature] = float(

        relative_change
    )

sensitivity_df = pd.DataFrame({

    "Feature": list(
        feature_sensitivity.keys()
    ),

    "Sensitivity": list(
        feature_sensitivity.values()
    )
})

sensitivity_df = sensitivity_df.sort_values(

    "Sensitivity",

    ascending=True
)

plt.figure(
    figsize=(11, 6)
)

bars = plt.barh(

    sensitivity_df["Feature"],

    sensitivity_df["Sensitivity"],

    color="#e74c3c",

    edgecolor="black"
)

for bar in bars:

    value = bar.get_width()

    plt.text(

        value + 0.02,

        bar.get_y() +
        bar.get_height() / 2,

        f"{value:.4f}%",

        va="center"
    )

plt.title(

    "PSO Model Feature Sensitivity Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel(

    "Mean Absolute Prediction Change (%)\n"
    "For a 1% Increase in Each Feature"
)

plt.xlim(

    0,

    max(

        1.2,

        sensitivity_df["Sensitivity"].max() * 1.20
    )
)

save_and_show(

    "pso_feature_importance_graph.png"
)

# ============================================================
# 36. PSO REGIONAL PERFORMANCE SUMMARY
# ============================================================

regional_summary = result_df.groupby(
    "region"
).agg(

    Average_Reported_Production=(

        "reported_sugar_production",

        "mean"
    ),

    Average_Baseline_Production=(

        "baseline_calculated_production",

        "mean"
    ),

    Average_PSO_Production=(

        "pso_optimized_production",

        "mean"
    ),

    Average_Additional_Production=(

        "additional_production",

        "mean"
    ),

    Average_Improvement_Percentage=(

        "improvement_percentage",

        "mean"
    ),

    Average_Original_Recovery=(

        "original_recovery_rate",

        "mean"
    ),

    Average_PSO_Recovery=(

        "pso_optimized_recovery_rate",

        "mean"
    )
)

regional_summary.round(8).to_csv(

    BASE_DIR / "pso_regional_summary.csv"
)

print("\nPSO Regional Summary:")

print(regional_summary.round(4))

# ============================================================
# 37. PSO PRODUCTION IMPROVEMENT GRAPH
# ============================================================

fig, ax = plt.subplots(
    figsize=(13, 7)
)

labels = (

    result_df["region"] +

    "\n" +

    result_df["season"]
)

positions = np.arange(
    len(result_df)
)

if "ais_improvement_percentage" in result_df:

    width = 0.35

    bars1 = ax.bar(

        positions - width / 2,

        result_df[
            "ais_improvement_percentage"
        ],

        width,

        color="#9b59b6",

        edgecolor="black",

        label="AIS"
    )

    bars2 = ax.bar(

        positions + width / 2,

        result_df[
            "improvement_percentage"
        ],

        width,

        color="#e74c3c",

        edgecolor="black",

        label="PSO"
    )

    annotate_bars(

        ax,

        bars1,

        decimals=4,

        suffix="%"
    )

    annotate_bars(

        ax,

        bars2,

        decimals=4,

        suffix="%"
    )

    ax.legend()

else:

    bars = ax.bar(

        positions,

        result_df[
            "improvement_percentage"
        ],

        color="#e74c3c",

        edgecolor="black"
    )

    annotate_bars(

        ax,

        bars,

        decimals=4,

        suffix="%"
    )

ax.set_xticks(positions)

ax.set_xticklabels(labels)

ax.set_title(

    "PSO Hypothetical Sugar Production Improvement",

    fontsize=16,

    fontweight="bold"
)

ax.set_ylabel(

    "Production Improvement (%)"
)

ax.margins(y=0.20)

ax.grid(

    axis="y",

    alpha=0.3
)

save_and_show(

    "pso_improvement_graph.png"
)

# ============================================================
# 38. SAVE YAML CONFIGURATION
# ============================================================

limitations = [

    "The original dataset contains four region-season observations.",

    "Training data is mathematically simulated.",

    "Original observations are not independent external validation.",

    "Tamil Nadu is included in All India figures.",

    "All India 2017-18 figures are provisional.",

    "PSO regression uses least-squares warm-start initialization.",

    "Estimation score is 100 minus MAPE, not classification accuracy.",

    "Production optimization constraints are hypothetical.",

    "The mathematical formula exactly defines simulated production.",

    "PSO and AIS may use different numbers of objective evaluations.",

    "Equal iteration counts do not establish equal computational budgets.",

    "Scenario improvements are not verified industrial improvements.",

    "Some comparison scenarios may exceed the intervention budget."
]

config = {

    "project": {

        "name": (

            "PSO-Based Sugar Industry Performance "
            "Analysis and Production Efficiency"
        ),

        "version": "1.0.0",

        "algorithm": "Particle Swarm Optimization",

        "created_at": datetime.now().isoformat()
    },

    "dataset": {

        "filename": DATASET_PATH.name,

        "path": str(DATASET_PATH),

        "original_shape": list(
            raw_df.shape
        ),

        "processed_observations": int(
            len(df)
        ),

        "simulated_samples": N_SIMULATED_SAMPLES,

        "features": FEATURES,

        "polynomial_features": POLYNOMIAL_FEATURES,

        "target": TARGET,

        "target_unit": "Lakh Metric Tonnes",

        "training_data_type": "Mathematical Simulation"
    },

    "pso_parameters": {

        "swarm_size": SWARM_SIZE,

        "iterations": ITERATIONS,

        "maximum_inertia": W_MAX,

        "minimum_inertia": W_MIN,

        "cognitive_coefficient": C1,

        "social_coefficient": C2,

        "velocity_limit": VELOCITY_LIMIT,

        "regularization": REGULARIZATION,

        "particle_dimensions": N_PARAMETERS,

        "initialization": "Least-squares warm start",

        "random_seed": SEED,

        "objective_evaluations": int(

            pso_training[
                "objective_evaluations"
            ]
        )
    },

    "training": {

        "training_samples": int(
            len(X_train)
        ),

        "validation_samples": int(
            len(X_val)
        ),

        "testing_samples": int(
            len(X_test)
        ),

        "original_observations": int(
            len(X_real)
        ),

        "target_mean": y_mean,

        "target_std": y_std
    },

    "production_optimization": {

        "maximum_factory_change": MAX_FACTORY_CHANGE,

        "maximum_cane_change": MAX_CANE_CHANGE,

        "maximum_recovery_change": MAX_RECOVERY_CHANGE,

        "maximum_recovery_rate": MAX_RECOVERY_RATE,

        "intervention_budget": INTERVENTION_BUDGET,

        "optimization_iterations": OPTIMIZATION_ITERATIONS,

        "maximum_relative_cane_per_factory": 1.10,

        "objective": (

            "Maximize formula-based sugar production "
            "under hypothetical intervention and "
            "processing constraints"
        )
    },

    "output_files": {

        "h5": H5_PATH.name,

        "pkl": PKL_PATH.name,

        "yaml": YAML_PATH.name,

        "json": JSON_PATH.name,

        "prefix": PREFIX
    },

    "limitations": limitations
}

with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        config,

        file,

        sort_keys=False,

        default_flow_style=False,

        allow_unicode=True
    )

print("\n[SAVED]", YAML_PATH.name)

# ============================================================
# 39. SAVE JSON METRICS
# ============================================================

# Convert NumPy values and missing values into
# JSON-compatible Python objects.

def json_safe(value):

    if isinstance(value, dict):

        return {

            str(key): json_safe(item)

            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):

        return [

            json_safe(item)

            for item in value
        ]

    if isinstance(value, np.ndarray):

        return json_safe(
            value.tolist()
        )

    if isinstance(value, np.generic):

        return json_safe(
            value.item()
        )

    if isinstance(value, float):

        return value if np.isfinite(value) else None

    if value is pd.NA:

        return None

    return value

results = {

    "project_name": config["project"]["name"],

    "algorithm": "Particle Swarm Optimization",

    "generated_at": datetime.now().isoformat(),

    "dataset_information": config["dataset"],

    "pso_parameters": config["pso_parameters"],

    "pso_metrics": pso_metrics,

    "baseline_metrics": baseline_metrics,

    "original_data_comparison": comparison_results,

    "simulation_test_comparison": (
        simulation_comparison_results
    ),

    "best_fitness": float(

        pso_training["best_fitness"]
    ),

    "best_particle": best_weights.tolist(),

    "training_history": training_history,

    "production_optimization_history": (
        optimization_histories
    ),

    "production_objective_evaluations": (
        optimization_evaluations
    ),

    "feature_sensitivity": feature_sensitivity,

    "optimization_results": (

        result_df.to_dict(
            orient="records"
        )
    ),

    "scenario_predictions": (

        prediction_df.to_dict(
            orient="records"
        )
    ),

    "regional_summary": (

        regional_summary.to_dict(
            orient="index"
        )
    ),

    "h5_verification_maximum_difference": (
        maximum_difference
    ),

    "limitations": limitations
}

with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_safe(results),

        file,

        indent=4,

        ensure_ascii=False,

        allow_nan=False
    )

print("\n[SAVED]", JSON_PATH.name)

# ============================================================
# 40. VERIFY SAVED PKL AND H5 MODELS
# ============================================================

loaded_package = joblib.load(
    PKL_PATH
)

sample_input = np.array([

    [37, 81.43, 9.50]

], dtype=float)

sample_poly = polynomial_transform(

    sample_input
)

sample_scaled = loaded_package[

    "scaler"

].transform(sample_poly)

sample_bias = add_bias(

    sample_scaled
)

sample_pkl_prediction = np.maximum(

    (

        sample_bias @

        loaded_package["weights"]

    ) *

    loaded_package["target_std"] +

    loaded_package["target_mean"],

    0.0
)

sample_h5_prediction = loaded_h5_model.predict(

    sample_input.astype(np.float32),

    verbose=0

).flatten()

sample_formula_prediction = (

    sample_input[0, 1] *

    sample_input[0, 2] / 100
)

print("\n" + "=" * 80)
print("SAVED MODEL VERIFICATION")
print("=" * 80)

print("\nSample Input:")

print(

    pd.DataFrame(

        sample_input,

        columns=FEATURES
    )
)

print(

    f"\nMathematical Prediction: "
    f"{sample_formula_prediction:.6f} Lakh MT"
)

print(

    f"PSO PKL Prediction: "
    f"{sample_pkl_prediction[0]:.6f} Lakh MT"
)

print(

    f"PSO H5 Prediction: "
    f"{sample_h5_prediction[0]:.6f} Lakh MT"
)

# ============================================================
# 41. FINAL PSO PERFORMANCE SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL PSO PERFORMANCE SUMMARY")
print("=" * 80)

print("\nPSO SIMULATION TEST PERFORMANCE")

for name, value in pso_metrics[
    "simulation_test"
].items():

    if value is not None:

        print(f"{name}: {value:.8f}")

print("\nPSO ORIGINAL DATA PERFORMANCE")

for name, value in pso_metrics[
    "original_observations"
].items():

    if value is not None:

        print(f"{name}: {value:.8f}")

print("\nREGIONAL PRODUCTION IMPROVEMENTS")

for region, regional_data in result_df.groupby(
    "region"
):

    average_improvement = (

        regional_data[
            "improvement_percentage"
        ].mean()
    )

    print(

        f"{region}: "
        f"{average_improvement:.4f}% "
        f"average hypothetical improvement"
    )

print("\nBest Model by Original Observation RMSE:")

print(

    comparison_df["RMSE"].idxmin()
)

print("\nBest Model by Original Observation MAPE:")

print(

    comparison_df["MAPE_Percentage"].idxmin()
)

print("\nPSO Training Objective Evaluations:")

print(

    pso_training["objective_evaluations"]
)

# ============================================================
# 42. VERIFY ALL GENERATED FILES
# ============================================================

expected_files = [

    "pso_sugar_production_model.h5",

    "pso_sugar_production_model.pkl",

    "pso_config.yaml",

    "pso_metrics.json",

    "pso_accuracy_graph.png",

    "pso_heatmap.png",

    "pso_comparison_graph.png",

    "pso_result.csv",

    "pso_result_graph.png",

    "pso_prediction.csv",

    "pso_prediction_graph.png",

    "pso_convergence_graph.png",

    "pso_optimization_graph.png",

    "pso_model_comparison.csv",

    "pso_simulation_comparison.csv",

    "pso_feature_importance_graph.png",

    "pso_regional_summary.csv",

    "pso_improvement_graph.png"
]

print("\n" + "=" * 80)
print("GENERATED PSO OUTPUT FILES")
print("=" * 80)

success_count = 0

for filename in expected_files:

    file_path = BASE_DIR / filename

    if file_path.exists():

        size_kb = (

            file_path.stat().st_size / 1024
        )

        print(

            f"[SUCCESS] {filename} "

            f"({size_kb:.2f} KB)"
        )

        success_count += 1

    else:

        print(f"[MISSING] {filename}")

print(

    f"\nTotal Generated Files: "

    f"{success_count}/{len(expected_files)}"
)

print("\nOutput Directory:")

print(BASE_DIR)

print("\n" + "=" * 80)
print("EXPERIMENTAL LIMITATIONS")
print("=" * 80)

for index, limitation in enumerate(

    limitations,

    start=1
):

    print(f"{index}. {limitation}")

print("\n" + "=" * 80)

if success_count == len(expected_files):

    print("PSO PROJECT COMPLETED SUCCESSFULLY")

else:

    print("PSO PROJECT COMPLETED WITH MISSING OUTPUTS")

print("=" * 80)
```
