```python
# ============================================================
# AIS-BASED SUGAR INDUSTRY PERFORMANCE ANALYSIS
# AND PRODUCTION EFFICIENCY OPTIMIZATION
#
# Algorithm: Artificial Immune System (CLONALG)
# All output files use the ais_ prefix
# ============================================================

import json
import yaml
import joblib
import random
import warnings
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf

from pathlib import Path
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.dpi"] = 120
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["font.size"] = 11

PREFIX = "ais_"

POPULATION_SIZE = 60
GENERATIONS = 200
SELECTED_ANTIBODIES = 12
CLONE_FACTOR = 8
MUTATION_RATE = 0.15
REPLACEMENT_RATE = 0.15

N_SIMULATED_SAMPLES = 6000

# ============================================================
# 2. FILE PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Sugar Industry Performance Analysis and Production Efficiency"
)

DATASET_PATH = BASE_DIR / (
    "performance_of_sugar_industry_sugar_season_2019.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

H5_PATH = BASE_DIR / "ais_sugar_production_model.h5"
PKL_PATH = BASE_DIR / "ais_sugar_production_model.pkl"
YAML_PATH = BASE_DIR / "ais_config.yaml"
JSON_PATH = BASE_DIR / "ais_metrics.json"

if not DATASET_PATH.exists():

    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 80)
print("AIS-BASED SUGAR INDUSTRY PERFORMANCE ANALYSIS")
print("AND PRODUCTION EFFICIENCY OPTIMIZATION")
print("=" * 80)

# ============================================================
# 3. LOAD ORIGINAL DATASET
# ============================================================

raw_df = pd.read_csv(DATASET_PATH)

raw_df.columns = raw_df.columns.str.strip()

raw_df["Parameters"] = (
    raw_df["Parameters"]
    .astype(str)
    .str.strip()
)

print("\nOriginal Dataset Shape:", raw_df.shape)

print("\nOriginal Dataset:")
print(raw_df.to_string(index=False))

# ============================================================
# 4. TRANSFORM DATASET
# ============================================================

parameter_mapping = {

    "Number of Factories Working":
        "working_factories",

    "Installed Crushing Capacity (in Lakh Tonnes) per day":
        "installed_crushing_capacity",

    "Cane Crushed (in Lakh M.T.)":
        "cane_crushed",

    "Sugar Production from Cane (in Lakh M.T.)":
        "sugar_production",

    "Recovery Rate (in Percentage)":
        "recovery_rate"
}

df = raw_df.set_index("Parameters")

df = df.drop(
    columns=["S.No"],
    errors="ignore"
)

df = df.rename(index=parameter_mapping)

df = df.T.reset_index()

df = df.rename(
    columns={"index": "region_season"}
)

df.columns.name = None

required_columns = [
    "working_factories",
    "installed_crushing_capacity",
    "cane_crushed",
    "sugar_production",
    "recovery_rate"
]

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:

    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

# ============================================================
# 5. REGION AND SEASON EXTRACTION
# ============================================================

df["region"] = np.where(

    df["region_season"].str.contains(
        "Tamil Nadu",
        case=False
    ),

    "Tamil Nadu",

    "All India"
)

df["season"] = df["region_season"].str.extract(
    r"(\d{4}-\d{2})"
)

df["year"] = pd.to_numeric(

    df["season"].str[:4],

    errors="coerce"
)

df["provisional"] = (
    df["region_season"]
    .str.contains(
        "Provisional",
        case=False
    )
)

for column in required_columns:

    df[column] = pd.to_numeric(

        df[column]
        .astype(str)
        .str.replace(",", "", regex=False),

        errors="coerce"
    )

df = df.dropna(

    subset=[
        "working_factories",
        "cane_crushed",
        "sugar_production",
        "recovery_rate"
    ]

).reset_index(drop=True)

df["label"] = (
    df["region"] +
    "\n" +
    df["season"]
)

print("\nProcessed Dataset Shape:", df.shape)

# ============================================================
# 6. FEATURE ENGINEERING
# ============================================================

df["production_per_factory"] = (

    df["sugar_production"] /
    df["working_factories"]
)

df["cane_per_factory"] = (

    df["cane_crushed"] /
    df["working_factories"]
)

df["effective_recovery_rate"] = (

    df["sugar_production"] /
    df["cane_crushed"]

) * 100

df["recovery_gap"] = (

    df["effective_recovery_rate"] -
    df["recovery_rate"]
)

df["mathematical_prediction"] = (

    df["cane_crushed"] *
    df["recovery_rate"] / 100
)

print("\nProcessed Dataset:")
print(df.to_string(index=False))

# ============================================================
# 7. GENERATE SIMULATED TRAINING DATA
# ============================================================

# Simulation formula:
#
# Sugar Production = Cane Crushed * Recovery Rate / 100
#
# These records are hypothetical and are not additional
# real-world sugar industry observations.

simulation_records = []

for i in range(N_SIMULATED_SAMPLES):

    reference = df.iloc[
        rng.integers(0, len(df))
    ]

    factories = max(

        1,

        int(
            round(
                reference["working_factories"] *
                rng.uniform(0.80, 1.20)
            )
        )
    )

    cane_per_factory = (

        reference["cane_crushed"] /
        reference["working_factories"]
    )

    cane_crushed = (

        factories *
        cane_per_factory *
        rng.uniform(0.70, 1.30)
    )

    recovery_rate = np.clip(

        reference["recovery_rate"] *
        rng.uniform(0.85, 1.15),

        6.0,

        13.0
    )

    sugar_production = (

        cane_crushed *
        recovery_rate / 100
    )

    simulation_records.append({

        "working_factories": factories,

        "cane_crushed": cane_crushed,

        "recovery_rate": recovery_rate,

        "sugar_production": sugar_production
    })

simulated_df = pd.DataFrame(
    simulation_records
)

print(
    "\nSimulated Dataset Shape:",
    simulated_df.shape
)

# ============================================================
# 8. FEATURE AND TARGET SELECTION
# ============================================================

FEATURES = [

    "working_factories",

    "cane_crushed",

    "recovery_rate"
]

TARGET = "sugar_production"

X = simulated_df[FEATURES].to_numpy(
    dtype=np.float64
)

y = simulated_df[TARGET].to_numpy(
    dtype=np.float64
)

X_real = df[FEATURES].to_numpy(
    dtype=np.float64
)

y_real = df[TARGET].to_numpy(
    dtype=np.float64
)

# ============================================================
# 9. TRAIN, VALIDATION AND TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(

        X_train_full,

        y_train_full,

        test_size=0.20,

        random_state=SEED
    )
)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))
print("Original Observations:", len(X_real))

# ============================================================
# 10. POLYNOMIAL FEATURE TRANSFORMATION
# ============================================================

# The interaction terms allow the AIS model to learn
# relationships between cane crushed and recovery rate.

POLYNOMIAL_FEATURES = [

    "working_factories",

    "cane_crushed",

    "recovery_rate",

    "factories_squared",

    "cane_squared",

    "recovery_squared",

    "factories_x_cane",

    "factories_x_recovery",

    "cane_x_recovery"
]

def polynomial_transform(data):

    data = np.asarray(
        data,
        dtype=np.float64
    )

    f = data[:, 0]
    c = data[:, 1]
    r = data[:, 2]

    return np.column_stack([

        f,

        c,

        r,

        f ** 2,

        c ** 2,

        r ** 2,

        f * c,

        f * r,

        c * r
    ])

X_train_poly = polynomial_transform(X_train)
X_val_poly = polynomial_transform(X_val)
X_test_poly = polynomial_transform(X_test)
X_real_poly = polynomial_transform(X_real)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train_poly
)

X_val_scaled = scaler.transform(
    X_val_poly
)

X_test_scaled = scaler.transform(
    X_test_poly
)

X_real_scaled = scaler.transform(
    X_real_poly
)

# Target normalization

y_mean = float(np.mean(y_train))
y_std = float(np.std(y_train))

if y_std == 0:

    y_std = 1.0

y_train_scaled = (

    y_train - y_mean

) / y_std

y_val_scaled = (

    y_val - y_mean

) / y_std

# Add bias column

def add_bias(data):

    return np.column_stack([

        data,

        np.ones(len(data))
    ])

X_train_bias = add_bias(
    X_train_scaled
)

X_val_bias = add_bias(
    X_val_scaled
)

X_test_bias = add_bias(
    X_test_scaled
)

X_real_bias = add_bias(
    X_real_scaled
)

N_PARAMETERS = X_train_bias.shape[1]

print("\nAIS Antibody Dimensions:", N_PARAMETERS)

# ============================================================
# 11. ARTIFICIAL IMMUNE SYSTEM IMPLEMENTATION
# ============================================================

def clonal_selection(

    objective,

    lower_bounds,

    upper_bounds,

    population_size=60,

    generations=200,

    selected_antibodies=12,

    clone_factor=8,

    mutation_rate=0.15,

    replacement_rate=0.15,

    initial_antibodies=None,

    seed=42,

    verbose=True
):

    local_rng = np.random.default_rng(seed)

    lower_bounds = np.asarray(
        lower_bounds,
        dtype=float
    )

    upper_bounds = np.asarray(
        upper_bounds,
        dtype=float
    )

    dimensions = len(lower_bounds)

    search_range = (
        upper_bounds - lower_bounds
    )

    population = local_rng.uniform(

        lower_bounds,

        upper_bounds,

        size=(
            population_size,
            dimensions
        )
    )

    if initial_antibodies is not None:

        initial_antibodies = np.atleast_2d(
            initial_antibodies
        )

        n_initial = min(

            len(initial_antibodies),

            population_size
        )

        population[:n_initial] = np.clip(

            initial_antibodies[:n_initial],

            lower_bounds,

            upper_bounds
        )

    best_antibody = None

    best_fitness = float("inf")

    history = []

    # ========================================================
    # CLONAL SELECTION LOOP
    # ========================================================

    for generation in range(generations):

        # Step 1: Evaluate antibody affinity

        fitness = np.array([

            objective(antibody)

            for antibody in population
        ])

        sorted_indices = np.argsort(
            fitness
        )

        population = population[
            sorted_indices
        ]

        fitness = fitness[
            sorted_indices
        ]

        # Step 2: Preserve the best antibody

        if fitness[0] < best_fitness:

            best_fitness = float(
                fitness[0]
            )

            best_antibody = (
                population[0].copy()
            )

        # Step 3: Select high-affinity antibodies

        selected = population[
            :selected_antibodies
        ]

        selected_fitness = fitness[
            :selected_antibodies
        ]

        fitness_min = selected_fitness.min()

        fitness_max = selected_fitness.max()

        fitness_range = max(

            fitness_max - fitness_min,

            1e-12
        )

        normalized_fitness = (

            selected_fitness - fitness_min

        ) / fitness_range

        clones = []

        progress = (

            generation /
            max(generations - 1, 1)
        )

        mutation_decay = (

            1.0 - 0.95 * progress
        )

        # Step 4: Cloning and hypermutation

        for rank, antibody in enumerate(
            selected
        ):

            number_of_clones = max(

                2,

                int(
                    clone_factor *
                    selected_antibodies /
                    (rank + 1)
                )
            )

            affinity = (

                1.0 -
                normalized_fitness[rank]
            )

            mutation_strength = (

                mutation_rate *

                mutation_decay *

                (0.05 + 0.95 * (1 - affinity))
            )

            for _ in range(number_of_clones):

                mutation = local_rng.normal(

                    loc=0.0,

                    scale=(
                        mutation_strength *
                        search_range
                    ),

                    size=dimensions
                )

                mutation_mask = (

                    local_rng.random(dimensions)

                    < 0.40
                )

                if not mutation_mask.any():

                    mutation_mask[
                        local_rng.integers(dimensions)
                    ] = True

                clone = (

                    antibody +

                    mutation * mutation_mask
                )

                clone = np.clip(

                    clone,

                    lower_bounds,

                    upper_bounds
                )

                clones.append(clone)

        clones = np.asarray(clones)

        # Step 5: Combine parents and mutated clones

        combined_population = np.vstack([

            population,

            clones,

            best_antibody.reshape(1, -1)
        ])

        combined_fitness = np.array([

            objective(antibody)

            for antibody in combined_population
        ])

        combined_indices = np.argsort(
            combined_fitness
        )

        population = combined_population[

            combined_indices[:population_size]
        ]

        # Update global memory

        current_best_index = (
            combined_indices[0]
        )

        current_best_fitness = (
            combined_fitness[current_best_index]
        )

        if current_best_fitness < best_fitness:

            best_fitness = float(
                current_best_fitness
            )

            best_antibody = (
                combined_population[
                    current_best_index
                ].copy()
            )

        # Step 6: Replace low-affinity antibodies
        # to maintain population diversity.

        replacement_count = max(

            1,

            int(
                population_size *
                replacement_rate
            )
        )

        population[-replacement_count:] = (
            local_rng.uniform(

                lower_bounds,

                upper_bounds,

                size=(
                    replacement_count,
                    dimensions
                )
            )
        )

        # Preserve immune memory

        population[0] = best_antibody.copy()

        history.append(
            float(best_fitness)
        )

        if verbose and (

            generation == 0 or

            (generation + 1) % 20 == 0 or

            generation == generations - 1
        ):

            print(

                f"Generation "
                f"{generation + 1:03d}/"
                f"{generations} | "

                f"Best Fitness: "
                f"{best_fitness:.10f}"
            )

    return {

        "best_antibody": best_antibody,

        "best_fitness": best_fitness,

        "history": history
    }

# ============================================================
# 12. AIS REGRESSION FITNESS FUNCTION
# ============================================================

# Each antibody represents regression coefficients
# and a bias parameter.
#
# Lower prediction error means higher affinity.

REGULARIZATION = 1e-8

def model_fitness(antibody):

    train_prediction = (
        X_train_bias @ antibody
    )

    validation_prediction = (
        X_val_bias @ antibody
    )

    train_mse = np.mean(

        (
            y_train_scaled -
            train_prediction
        ) ** 2
    )

    validation_mse = np.mean(

        (
            y_val_scaled -
            validation_prediction
        ) ** 2
    )

    regularization_penalty = (

        REGULARIZATION *

        np.sum(antibody[:-1] ** 2)
    )

    fitness = (

        0.70 * train_mse +

        0.30 * validation_mse +

        regularization_penalty
    )

    return float(fitness)

# ============================================================
# 13. TRAIN AIS MODEL
# ============================================================

print("\n" + "=" * 80)
print("TRAINING AIS REGRESSION MODEL")
print("=" * 80)

lower_bounds = np.full(

    N_PARAMETERS,

    -5.0
)

upper_bounds = np.full(

    N_PARAMETERS,

    5.0
)

# Use least-squares initialization as one candidate.
# This is a warm-started AIS optimization, not a
# purely random initialization.

least_squares_weights = np.linalg.lstsq(

    X_train_bias,

    y_train_scaled,

    rcond=None

)[0]

ais_training = clonal_selection(

    objective=model_fitness,

    lower_bounds=lower_bounds,

    upper_bounds=upper_bounds,

    population_size=POPULATION_SIZE,

    generations=GENERATIONS,

    selected_antibodies=SELECTED_ANTIBODIES,

    clone_factor=CLONE_FACTOR,

    mutation_rate=MUTATION_RATE,

    replacement_rate=REPLACEMENT_RATE,

    initial_antibodies=[
        least_squares_weights
    ],

    seed=SEED,

    verbose=True
)

best_weights = ais_training[
    "best_antibody"
]

training_history = ais_training[
    "history"
]

print("\nBest AIS Fitness:")

print(
    ais_training["best_fitness"]
)

print("\nOptimized AIS Weights:")

print(best_weights)

# ============================================================
# 14. AIS PREDICTION FUNCTION
# ============================================================

def ais_predict(data):

    data = np.asarray(

        data,

        dtype=np.float64
    )

    polynomial_data = (
        polynomial_transform(data)
    )

    scaled_data = scaler.transform(
        polynomial_data
    )

    bias_data = add_bias(
        scaled_data
    )

    normalized_prediction = (

        bias_data @ best_weights
    )

    prediction = (

        normalized_prediction *
        y_std +

        y_mean
    )

    return np.maximum(

        prediction,

        0.0
    )

# ============================================================
# 15. GENERATE MODEL PREDICTIONS
# ============================================================

ais_train_predictions = ais_predict(
    X_train
)

ais_val_predictions = ais_predict(
    X_val
)

ais_test_predictions = ais_predict(
    X_test
)

ais_real_predictions = ais_predict(
    X_real
)

baseline_test_predictions = (

    X_test[:, 1] *

    X_test[:, 2] / 100
)

baseline_real_predictions = (

    X_real[:, 1] *

    X_real[:, 2] / 100
)

df["ais_prediction"] = (
    ais_real_predictions
)

df["ais_absolute_error"] = np.abs(

    df["sugar_production"] -

    df["ais_prediction"]
)

# ============================================================
# 16. MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(

        actual,

        dtype=float

    ).flatten()

    predicted = np.asarray(

        predicted,

        dtype=float

    ).flatten()

    mae = mean_absolute_error(

        actual,

        predicted
    )

    mse = mean_squared_error(

        actual,

        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(

        actual,

        predicted
    )

    nonzero = actual != 0

    if np.any(nonzero):

        mape = np.mean(

            np.abs(

                (
                    actual[nonzero] -

                    predicted[nonzero]
                ) /

                actual[nonzero]
            )

        ) * 100

        estimation_score = 100 - mape

    else:

        mape = None

        estimation_score = None

    return {

        "MAE": float(mae),

        "MSE": float(mse),

        "RMSE": float(rmse),

        "R2_Score": float(r2),

        "MAPE_Percentage": (

            float(mape)

            if mape is not None

            else None
        ),

        "Estimation_Score_Percentage": (

            float(estimation_score)

            if estimation_score is not None

            else None
        )
    }

# ============================================================
# 17. AIS EVALUATION
# ============================================================

ais_metrics = {

    "training": evaluate_model(

        y_train,

        ais_train_predictions
    ),

    "validation": evaluate_model(

        y_val,

        ais_val_predictions
    ),

    "simulation_test": evaluate_model(

        y_test,

        ais_test_predictions
    ),

    "original_observations": evaluate_model(

        y_real,

        ais_real_predictions
    )
}

baseline_metrics = {

    "simulation_test": evaluate_model(

        y_test,

        baseline_test_predictions
    ),

    "original_observations": evaluate_model(

        y_real,

        baseline_real_predictions
    )
}

print("\n" + "=" * 80)
print("AIS MODEL EVALUATION")
print("=" * 80)

for dataset_name, values in ais_metrics.items():

    print(f"\n{dataset_name.upper()}")

    for metric_name, value in values.items():

        if value is not None:

            print(
                f"{metric_name}: {value:.8f}"
            )

# ============================================================
# 18. LOAD PREVIOUS BASELINE MODEL RESULTS
# ============================================================

comparison_results = {

    "Mathematical Baseline":
        baseline_metrics["original_observations"]
}

simulation_comparison_results = {

    "Mathematical Baseline":
        baseline_metrics["simulation_test"]
}

previous_pkl = (
    BASE_DIR / "sugar_production_model.pkl"
)

previous_h5 = (
    BASE_DIR / "sugar_production_model.h5"
)

previous_json = (
    BASE_DIR / "metrics.json"
)

rf_model = None
ann_model = None

if previous_pkl.exists():

    try:

        rf_model = joblib.load(
            previous_pkl
        )

        rf_real_predictions = rf_model.predict(

            pd.DataFrame(
                X_real,
                columns=FEATURES
            )
        )

        rf_test_predictions = rf_model.predict(

            pd.DataFrame(
                X_test,
                columns=FEATURES
            )
        )

        comparison_results["Random Forest"] = (
            evaluate_model(
                y_real,
                rf_real_predictions
            )
        )

        simulation_comparison_results[
            "Random Forest"
        ] = evaluate_model(

            y_test,

            rf_test_predictions
        )

        df["random_forest_prediction"] = (
            rf_real_predictions
        )

        print("\nPrevious Random Forest model loaded.")

    except Exception as error:

        print(
            "\nRandom Forest loading error:",
            error
        )

        rf_model = None

if previous_h5.exists():

    try:

        ann_model = tf.keras.models.load_model(

            str(previous_h5),

            compile=False
        )

        ann_real_predictions = ann_model.predict(

            X_real.astype(np.float32),

            verbose=0

        ).flatten()

        ann_test_predictions = ann_model.predict(

            X_test.astype(np.float32),

            verbose=0

        ).flatten()

        comparison_results["ANN"] = (
            evaluate_model(

                y_real,

                ann_real_predictions
            )
        )

        simulation_comparison_results["ANN"] = (
            evaluate_model(

                y_test,

                ann_test_predictions
            )
        )

        df["ann_prediction"] = (
            ann_real_predictions
        )

        print("\nPrevious ANN model loaded.")

    except Exception as error:

        print(
            "\nANN loading error:",
            error
        )

        ann_model = None

comparison_results["AIS"] = (

    ais_metrics["original_observations"]
)

simulation_comparison_results["AIS"] = (

    ais_metrics["simulation_test"]
)

comparison_df = pd.DataFrame(
    comparison_results
).T

comparison_df.index.name = "Model"

simulation_comparison_df = pd.DataFrame(
    simulation_comparison_results
).T

simulation_comparison_df.index.name = "Model"

comparison_df.to_csv(

    BASE_DIR / "ais_model_comparison.csv"
)

simulation_comparison_df.to_csv(

    BASE_DIR / "ais_simulation_comparison.csv"
)

print("\nOriginal Observation Comparison:")

print(comparison_df.round(6))

print("\nSimulation Test Comparison:")

print(simulation_comparison_df.round(6))

# ============================================================
# 19. SAVE AIS MODEL AS PKL
# ============================================================

# Store standard Python and NumPy objects so the
# file can be loaded without a custom model class.

ais_model_package = {

    "algorithm": "Artificial Immune System",

    "method": "Clonal Selection Algorithm",

    "features": FEATURES,

    "polynomial_features": POLYNOMIAL_FEATURES,

    "scaler": scaler,

    "weights": best_weights,

    "target_mean": y_mean,

    "target_std": y_std,

    "training_data_type": "Simulated",

    "random_seed": SEED
}

joblib.dump(

    ais_model_package,

    PKL_PATH
)

print("\n[SAVED]", PKL_PATH.name)

# ============================================================
# 20. CONVERT AIS MODEL INTO KERAS H5
# ============================================================

# The H5 model reproduces the AIS regression equation.
# No backpropagation is used to optimize its weights.
#
# Input order:
# factories, cane crushed, recovery rate

inputs = tf.keras.Input(

    shape=(3,),

    name="raw_industry_features"
)

f = tf.keras.layers.Cropping1D(
    cropping=(0, 2)
)(
    tf.keras.layers.Reshape((3, 1))(inputs)
)

f = tf.keras.layers.Flatten()(f)

c = tf.keras.layers.Cropping1D(
    cropping=(1, 1)
)(
    tf.keras.layers.Reshape((3, 1))(inputs)
)

c = tf.keras.layers.Flatten()(c)

r = tf.keras.layers.Cropping1D(
    cropping=(2, 0)
)(
    tf.keras.layers.Reshape((3, 1))(inputs)
)

r = tf.keras.layers.Flatten()(r)

f_squared = tf.keras.layers.Multiply()(
    [f, f]
)

c_squared = tf.keras.layers.Multiply()(
    [c, c]
)

r_squared = tf.keras.layers.Multiply()(
    [r, r]
)

fc = tf.keras.layers.Multiply()(
    [f, c]
)

fr = tf.keras.layers.Multiply()(
    [f, r]
)

cr = tf.keras.layers.Multiply()(
    [c, r]
)

polynomial_layer = tf.keras.layers.Concatenate(

    name="polynomial_features"

)([

    f,

    c,

    r,

    f_squared,

    c_squared,

    r_squared,

    fc,

    fr,

    cr
])

# Fold feature normalization and target scaling
# into the final Dense layer.

raw_coefficients = (

    best_weights[:-1] *

    y_std /

    scaler.scale_
)

raw_bias = (

    best_weights[-1] * y_std +

    y_mean -

    np.sum(

        raw_coefficients *

        scaler.mean_
    )
)

output_layer = tf.keras.layers.Dense(

    1,

    activation="relu",

    name="ais_optimized_regression"
)

outputs = output_layer(
    polynomial_layer
)

ais_keras_model = tf.keras.Model(

    inputs=inputs,

    outputs=outputs,

    name="AIS_Sugar_Production_Model"
)

output_layer.set_weights([

    raw_coefficients.reshape(
        -1, 1
    ).astype(np.float32),

    np.array(
        [raw_bias],
        dtype=np.float32
    )
])

ais_keras_model.save(

    str(H5_PATH),

    include_optimizer=False
)

print("\n[SAVED]", H5_PATH.name)

# Verify H5 model equivalence

loaded_h5_model = tf.keras.models.load_model(

    str(H5_PATH),

    compile=False
)

h5_predictions = loaded_h5_model.predict(

    X_real.astype(np.float32),

    verbose=0

).flatten()

maximum_difference = float(

    np.max(

        np.abs(

            h5_predictions -

            ais_real_predictions
        )
    )
)

print(

    "\nMaximum difference between H5 and AIS model:",

    maximum_difference
)

# ============================================================
# 21. GRAPH SAVING FUNCTION
# ============================================================

def save_and_show(filename):

    if not filename.startswith(PREFIX):

        filename = PREFIX + filename

    plt.tight_layout()

    plt.savefig(

        BASE_DIR / filename,

        dpi=300,

        bbox_inches="tight",

        facecolor="white"
    )

    plt.show()

    plt.close()

    print(f"[SAVED] {filename}")

# ============================================================
# 22. AIS ACCURACY GRAPH
# ============================================================

dataset_names = [

    "Training",

    "Validation",

    "Simulation Test",

    "Original Observations"
]

accuracy_values = [

    ais_metrics["training"][
        "Estimation_Score_Percentage"
    ],

    ais_metrics["validation"][
        "Estimation_Score_Percentage"
    ],

    ais_metrics["simulation_test"][
        "Estimation_Score_Percentage"
    ],

    ais_metrics["original_observations"][
        "Estimation_Score_Percentage"
    ]
]

plt.figure(figsize=(12, 7))

bars = plt.bar(

    dataset_names,

    accuracy_values,

    color=[
        "#3498db",
        "#2ecc71",
        "#9b59b6",
        "#e67e22"
    ],

    edgecolor="black"
)

for bar in bars:

    height = bar.get_height()

    plt.annotate(

        f"{height:.4f}%",

        (
            bar.get_x() +
            bar.get_width() / 2,

            height
        ),

        xytext=(0, 5),

        textcoords="offset points",

        ha="center",

        fontweight="bold"
    )

plt.title(

    "AIS-Based Sugar Production Estimation Performance\n"
    "(100 - MAPE; Not Classification Accuracy)",

    fontsize=16,

    fontweight="bold"
)

plt.ylabel("Estimation Score (%)")

plt.ylim(

    min(0, min(accuracy_values) - 10),

    max(110, max(accuracy_values) + 10)
)

plt.grid(

    axis="y",

    alpha=0.3
)

save_and_show("ais_accuracy_graph.png")

# ============================================================
# 23. AIS CORRELATION HEATMAP
# ============================================================

heatmap_columns = [

    "working_factories",

    "installed_crushing_capacity",

    "cane_crushed",

    "sugar_production",

    "recovery_rate",

    "production_per_factory",

    "cane_per_factory",

    "effective_recovery_rate",

    "ais_prediction"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(14, 11))

sns.heatmap(

    correlation_matrix,

    annot=True,

    fmt=".2f",

    cmap="coolwarm",

    center=0,

    vmin=-1,

    vmax=1,

    linewidths=0.5,

    square=True,

    cbar_kws={

        "label": "Correlation Coefficient"
    }
)

plt.title(

    "AIS Sugar Industry Correlation Heatmap\n"
    "Descriptive Analysis of Original Observations",

    fontsize=16,

    fontweight="bold"
)

plt.xticks(

    rotation=45,

    ha="right"
)

plt.yticks(rotation=0)

save_and_show("ais_heatmap.png")

# ============================================================
# 24. AIS COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(

    2,

    2,

    figsize=(16, 11)
)

comparison_metrics = [

    "MAE",

    "RMSE",

    "R2_Score",

    "MAPE_Percentage"
]

metric_titles = [

    "Mean Absolute Error",

    "Root Mean Squared Error",

    "R² Score",

    "Mean Absolute Percentage Error"
]

for ax, metric, title in zip(

    axes.flatten(),

    comparison_metrics,

    metric_titles
):

    colors = [

        "#e74c3c"

        if model == "AIS"

        else "#3498db"

        for model in comparison_df.index
    ]

    bars = ax.bar(

        comparison_df.index,

        comparison_df[metric],

        color=colors,

        edgecolor="black"
    )

    for bar in bars:

        height = bar.get_height()

        ax.annotate(

            f"{height:.4f}",

            (
                bar.get_x() +
                bar.get_width() / 2,

                height
            ),

            xytext=(0, 5),

            textcoords="offset points",

            ha="center",

            fontsize=9
        )

    ax.set_title(

        title,

        fontweight="bold"
    )

    ax.tick_params(

        axis="x",

        rotation=15
    )

    ax.margins(y=0.20)

    ax.grid(

        axis="y",

        alpha=0.3
    )

fig.suptitle(

    "Comparative Performance Analysis Including AIS\n"
    "Original Region-Season Observations",

    fontsize=18,

    fontweight="bold"
)

save_and_show("ais_comparison_graph.png")

# ============================================================
# 25. AIS MODEL CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(

    np.arange(
        1,
        len(training_history) + 1
    ),

    np.maximum(
        training_history,
        1e-16
    ),

    color="#8e44ad",

    linewidth=2.5,

    label="Best AIS Fitness"
)

plt.yscale("log")

plt.xlabel("Generation")

plt.ylabel("Regularized Prediction Error")

plt.title(

    "AIS Clonal Selection Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.legend()

plt.grid(alpha=0.3)

save_and_show("ais_convergence_graph.png")

# ============================================================
# 26. AIS PRODUCTION OPTIMIZATION
# ============================================================

# Decision variables:
#
# x[0] = relative change in working factories
# x[1] = relative change in cane crushed
# x[2] = relative change in recovery rate
#
# Hypothetical constraints:
#
# Factory expansion: 0% to 5%
# Cane availability increase: 0% to 10%
# Recovery improvement: 0% to 5%
#
# A normalized intervention budget prevents every
# variable from simply increasing to its maximum.
#
# These are assumed experimental constraints,
# not measured industrial costs.

MAX_FACTORY_CHANGE = 0.05

MAX_CANE_CHANGE = 0.10

MAX_RECOVERY_CHANGE = 0.05

MAX_RECOVERY_RATE = 13.0

INTERVENTION_BUDGET = 1.50

OPTIMIZATION_GENERATIONS = 150

optimization_records = []

optimization_histories = {}

def decode_solution(reference, antibody):

    factory_change = float(
        antibody[0]
    )

    cane_change = float(
        antibody[1]
    )

    recovery_change = float(
        antibody[2]
    )

    factories = max(

        1,

        int(

            np.floor(

                reference["working_factories"] *

                (1 + factory_change)

                + 1e-9
            )
        )
    )

    cane_crushed = (

        reference["cane_crushed"] *

        (1 + cane_change)
    )

    recovery_rate = min(

        reference["recovery_rate"] *

        (1 + recovery_change),

        MAX_RECOVERY_RATE
    )

    return (

        factories,

        float(cane_crushed),

        float(recovery_rate)
    )

def intervention_cost(antibody):

    return float(

        antibody[0] / MAX_FACTORY_CHANGE +

        antibody[1] / MAX_CANE_CHANGE +

        antibody[2] / MAX_RECOVERY_CHANGE
    )

def repair_budget(antibody):

    antibody = np.asarray(

        antibody,

        dtype=float
    ).copy()

    cost = intervention_cost(
        antibody
    )

    if cost > INTERVENTION_BUDGET:

        antibody *= (

            INTERVENTION_BUDGET /

            cost
        )

    return antibody

for index, reference in df.iterrows():

    region = reference["region"]

    season = reference["season"]

    baseline_production = (

        reference["cane_crushed"] *

        reference["recovery_rate"] / 100
    )

    baseline_cane_per_factory = (

        reference["cane_crushed"] /

        reference["working_factories"]
    )

    def production_fitness(antibody):

        antibody = repair_budget(
            antibody
        )

        factories, cane, recovery = (
            decode_solution(

                reference,

                antibody
            )
        )

        production = (

            cane *

            recovery / 100
        )

        # Limit cane processed per factory to
        # 110% of the reference operating level.

        maximum_cane_per_factory = (

            baseline_cane_per_factory *

            1.10
        )

        capacity_violation = max(

            0.0,

            (
                cane / factories -

                maximum_cane_per_factory
            ) /

            maximum_cane_per_factory
        )

        production_gain = (

            production /

            baseline_production - 1
        )

        # Minimize negative production gain,
        # with a small intervention penalty.

        objective = (

            -production_gain +

            0.005 * intervention_cost(
                antibody
            ) +

            1000.0 * capacity_violation
        )

        return float(objective)

    print("\n" + "-" * 80)

    print(

        f"OPTIMIZING: {region} | {season}"
    )

    optimization = clonal_selection(

        objective=production_fitness,

        lower_bounds=[
            0.0,
            0.0,
            0.0
        ],

        upper_bounds=[

            MAX_FACTORY_CHANGE,

            MAX_CANE_CHANGE,

            MAX_RECOVERY_CHANGE
        ],

        population_size=50,

        generations=OPTIMIZATION_GENERATIONS,

        selected_antibodies=10,

        clone_factor=6,

        mutation_rate=0.20,

        replacement_rate=0.15,

        initial_antibodies=[
            [0.0, 0.0, 0.0]
        ],

        seed=SEED + index,

        verbose=False
    )

    best_solution = repair_budget(

        optimization["best_antibody"]
    )

    optimized_factories, optimized_cane, optimized_recovery = (
        decode_solution(

            reference,

            best_solution
        )
    )

    # Repair any residual capacity violation.

    maximum_allowed_cane = (

        optimized_factories *

        baseline_cane_per_factory *

        1.10
    )

    optimized_cane = min(

        optimized_cane,

        maximum_allowed_cane
    )

    best_solution[1] = max(

        0.0,

        optimized_cane /

        reference["cane_crushed"] - 1
    )

    optimized_production = (

        optimized_cane *

        optimized_recovery / 100
    )

    optimized_ais_prediction = float(

        ais_predict([[

            optimized_factories,

            optimized_cane,

            optimized_recovery

        ]])[0]
    )

    additional_production = (

        optimized_production -

        baseline_production
    )

    improvement_percentage = (

        additional_production /

        baseline_production

    ) * 100

    optimization_records.append({

        "region": region,

        "season": season,

        "provisional": bool(
            reference["provisional"]
        ),

        "original_factories": int(
            reference["working_factories"]
        ),

        "ais_optimized_factories": int(
            optimized_factories
        ),

        "original_cane_crushed": float(
            reference["cane_crushed"]
        ),

        "ais_optimized_cane_crushed": float(
            optimized_cane
        ),

        "original_recovery_rate": float(
            reference["recovery_rate"]
        ),

        "ais_optimized_recovery_rate": float(
            optimized_recovery
        ),

        "reported_sugar_production": float(
            reference["sugar_production"]
        ),

        "baseline_calculated_production": float(
            baseline_production
        ),

        "ais_original_prediction": float(
            reference["ais_prediction"]
        ),

        "ais_optimized_production": float(
            optimized_production
        ),

        "ais_optimized_prediction": float(
            optimized_ais_prediction
        ),

        "additional_production": float(
            additional_production
        ),

        "improvement_percentage": float(
            improvement_percentage
        ),

        "intervention_budget_used": float(
            intervention_cost(best_solution)
        ),

        "optimization_fitness": float(
            production_fitness(best_solution)
        ),

        "data_type": "Hypothetical Optimization"
    })

    optimization_histories[

        f"{region} {season}"

    ] = optimization["history"]

    print(

        f"Baseline Production: "
        f"{baseline_production:.4f} Lakh MT"
    )

    print(

        f"AIS Optimized Production: "
        f"{optimized_production:.4f} Lakh MT"
    )

    print(

        f"Production Improvement: "
        f"{improvement_percentage:.4f}%"
    )

# ============================================================
# 27. SAVE AIS RESULT CSV
# ============================================================

result_df = pd.DataFrame(
    optimization_records
)

result_df["original_production_per_factory"] = (

    result_df["reported_sugar_production"] /

    result_df["original_factories"]
)

result_df["ais_production_per_factory"] = (

    result_df["ais_optimized_production"] /

    result_df["ais_optimized_factories"]
)

result_df.round(8).to_csv(

    BASE_DIR / "ais_result.csv",

    index=False
)

print("\n" + "=" * 80)
print("AIS OPTIMIZATION RESULTS")
print("=" * 80)

print(

    result_df.round(4).to_string(
        index=False
    )
)

print("\n[SAVED] ais_result.csv")

# ============================================================
# 28. AIS RESULT GRAPH
# ============================================================

regions = result_df["region"].unique()

fig, axes = plt.subplots(

    1,

    len(regions),

    figsize=(16, 7),

    squeeze=False
)

for ax, region in zip(

    axes.flatten(),

    regions
):

    regional_data = result_df[

        result_df["region"] == region

    ].sort_values("season")

    positions = np.arange(
        len(regional_data)
    )

    width = 0.20

    columns = [

        "reported_sugar_production",

        "baseline_calculated_production",

        "ais_original_prediction",

        "ais_optimized_production"
    ]

    labels = [

        "Reported Production",

        "Mathematical Baseline",

        "AIS Prediction",

        "AIS Optimized Production"
    ]

    colors = [

        "#34495e",

        "#3498db",

        "#9b59b6",

        "#2ecc71"
    ]

    for i, (column, label, color) in enumerate(

        zip(
            columns,
            labels,
            colors
        )
    ):

        bars = ax.bar(

            positions + (i - 1.5) * width,

            regional_data[column],

            width,

            label=label,

            color=color,

            edgecolor="black"
        )

        for bar in bars:

            height = bar.get_height()

            ax.annotate(

                f"{height:.2f}",

                (
                    bar.get_x() +
                    bar.get_width() / 2,

                    height
                ),

                xytext=(0, 4),

                textcoords="offset points",

                ha="center",

                rotation=90,

                fontsize=8
            )

    ax.set_xticks(positions)

    ax.set_xticklabels(
        regional_data["season"]
    )

    ax.set_title(

        region,

        fontweight="bold"
    )

    ax.set_ylabel(
        "Sugar Production (Lakh MT)"
    )

    ax.margins(y=0.25)

    ax.legend(fontsize=8)

fig.suptitle(

    "AIS-Based Sugar Production Efficiency Optimization",

    fontsize=18,

    fontweight="bold"
)

save_and_show("ais_result_graph.png")

# ============================================================
# 29. PRODUCTION OPTIMIZATION CONVERGENCE
# ============================================================

plt.figure(figsize=(13, 8))

for label, history in optimization_histories.items():

    plt.plot(

        np.arange(
            1,
            len(history) + 1
        ),

        history,

        linewidth=2,

        label=label
    )

plt.title(

    "AIS Production Optimization Convergence",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel("Generation")

plt.ylabel("Optimization Objective (Lower is Better)")

plt.legend()

plt.grid(alpha=0.3)

save_and_show("ais_optimization_graph.png")

# ============================================================
# 30. GENERATE AIS PREDICTION SCENARIOS
# ============================================================

latest_df = (

    df.sort_values("year")

    .groupby("region")

    .tail(1)

    .reset_index(drop=True)
)

scenarios = [

    {
        "scenario": "Current Conditions",
        "factory_change": 0.00,
        "cane_change": 0.00,
        "recovery_change": 0.00
    },

    {
        "scenario": "Recovery Improvement",
        "factory_change": 0.00,
        "cane_change": 0.00,
        "recovery_change": 0.05
    },

    {
        "scenario": "Increased Cane Availability",
        "factory_change": 0.00,
        "cane_change": 0.10,
        "recovery_change": 0.00
    },

    {
        "scenario": "Factory Expansion",
        "factory_change": 0.05,
        "cane_change": 0.05,
        "recovery_change": 0.00
    },

    {
        "scenario": "Combined Improvement",
        "factory_change": 0.05,
        "cane_change": 0.10,
        "recovery_change": 0.05
    },

    {
        "scenario": "Reduced Cane Availability",
        "factory_change": 0.00,
        "cane_change": -0.10,
        "recovery_change": 0.00
    }
]

prediction_records = []

for _, reference in latest_df.iterrows():

    baseline_production = (

        reference["cane_crushed"] *

        reference["recovery_rate"] / 100
    )

    for scenario in scenarios:

        factories = max(

            1,

            int(

                np.floor(

                    reference["working_factories"] *

                    (1 + scenario["factory_change"])

                    + 1e-9
                )
            )
        )

        cane = (

            reference["cane_crushed"] *

            (1 + scenario["cane_change"])
        )

        recovery = np.clip(

            reference["recovery_rate"] *

            (1 + scenario["recovery_change"]),

            6.0,

            MAX_RECOVERY_RATE
        )

        prediction_records.append({

            "region": reference["region"],

            "reference_season": reference["season"],

            "scenario": scenario["scenario"],

            "working_factories": factories,

            "cane_crushed": float(cane),

            "recovery_rate": float(recovery),

            "reference_reported_production": float(
                reference["sugar_production"]
            ),

            "reference_calculated_production": float(
                baseline_production
            ),

            "data_type": "Hypothetical Scenario"
        })

    # Include the AIS optimized operating condition.

    optimized_row = result_df[

        (
            result_df["region"] ==
            reference["region"]
        )

        &

        (
            result_df["season"] ==
            reference["season"]
        )

    ].iloc[0]

    prediction_records.append({

        "region": reference["region"],

        "reference_season": reference["season"],

        "scenario": "AIS Optimized Scenario",

        "working_factories": int(

            optimized_row[
                "ais_optimized_factories"
            ]
        ),

        "cane_crushed": float(

            optimized_row[
                "ais_optimized_cane_crushed"
            ]
        ),

        "recovery_rate": float(

            optimized_row[
                "ais_optimized_recovery_rate"
            ]
        ),

        "reference_reported_production": float(

            reference["sugar_production"]
        ),

        "reference_calculated_production": float(

            baseline_production
        ),

        "data_type": "AIS Hypothetical Optimization"
    })

prediction_df = pd.DataFrame(
    prediction_records
)

# ============================================================
# 31. AIS SCENARIO PREDICTION
# ============================================================

X_future = prediction_df[
    FEATURES
].to_numpy(dtype=float)

prediction_df["mathematical_prediction"] = (

    prediction_df["cane_crushed"] *

    prediction_df["recovery_rate"] / 100
)

prediction_df["ais_prediction"] = (
    ais_predict(X_future)
)

if rf_model is not None:

    prediction_df["random_forest_prediction"] = (

        rf_model.predict(

            prediction_df[FEATURES]
        )
    )

if ann_model is not None:

    prediction_df["ann_prediction"] = (

        ann_model.predict(

            X_future.astype(np.float32),

            verbose=0

        ).flatten()
    )

prediction_df["ais_expected_change"] = (

    prediction_df["ais_prediction"] -

    prediction_df["reference_calculated_production"]
)

prediction_df["ais_improvement_percentage"] = (

    prediction_df["ais_expected_change"] /

    prediction_df["reference_calculated_production"]

) * 100

prediction_df["formula_improvement_percentage"] = (

    (

        prediction_df["mathematical_prediction"] -

        prediction_df["reference_calculated_production"]
    ) /

    prediction_df["reference_calculated_production"]

) * 100

prediction_df["performance_category"] = np.select(

    [

        prediction_df[
            "formula_improvement_percentage"
        ] >= 10,

        prediction_df[
            "formula_improvement_percentage"
        ] > 1e-8,

        prediction_df[
            "formula_improvement_percentage"
        ] < -1e-8
    ],

    [

        "High Improvement",

        "Moderate Improvement",

        "Production Decline"
    ],

    default="Stable"
)

# Check each scenario against the AIS intervention budget.

scenario_budget_values = []

for _, row in prediction_df.iterrows():

    reference = latest_df[

        latest_df["region"] ==
        row["region"]

    ].iloc[0]

    cost = (

        max(

            0,

            row["working_factories"] /

            reference["working_factories"] - 1

        ) / MAX_FACTORY_CHANGE

        +

        max(

            0,

            row["cane_crushed"] /

            reference["cane_crushed"] - 1

        ) / MAX_CANE_CHANGE

        +

        max(

            0,

            row["recovery_rate"] /

            reference["recovery_rate"] - 1

        ) / MAX_RECOVERY_CHANGE
    )

    scenario_budget_values.append(
        float(cost)
    )

prediction_df["intervention_budget_used"] = (
    scenario_budget_values
)

prediction_df["within_ais_budget"] = (

    prediction_df["intervention_budget_used"]

    <= INTERVENTION_BUDGET + 1e-8
)

# ============================================================
# 32. SAVE AIS PREDICTION CSV
# ============================================================

prediction_df.round(8).to_csv(

    BASE_DIR / "ais_prediction.csv",

    index=False
)

print("\n" + "=" * 80)
print("AIS SUGAR PRODUCTION PREDICTIONS")
print("=" * 80)

print(

    prediction_df.round(4).to_string(
        index=False
    )
)

print("\n[SAVED] ais_prediction.csv")

# ============================================================
# 33. AIS PREDICTION GRAPH
# ============================================================

regions = prediction_df["region"].unique()

fig, axes = plt.subplots(

    len(regions),

    1,

    figsize=(16, 7 * len(regions)),

    squeeze=False
)

for ax, region in zip(

    axes.flatten(),

    regions
):

    regional_predictions = prediction_df[

        prediction_df["region"] == region

    ].reset_index(drop=True)

    positions = np.arange(
        len(regional_predictions)
    )

    ax.plot(

        positions,

        regional_predictions[
            "mathematical_prediction"
        ],

        marker="o",

        linewidth=2.5,

        label="Mathematical Baseline"
    )

    if "random_forest_prediction" in (
        regional_predictions.columns
    ):

        ax.plot(

            positions,

            regional_predictions[
                "random_forest_prediction"
            ],

            marker="s",

            linewidth=2,

            label="Random Forest"
        )

    if "ann_prediction" in (
        regional_predictions.columns
    ):

        ax.plot(

            positions,

            regional_predictions[
                "ann_prediction"
            ],

            marker="^",

            linewidth=2,

            label="ANN"
        )

    ax.plot(

        positions,

        regional_predictions[
            "ais_prediction"
        ],

        marker="D",

        color="#8e44ad",

        linewidth=3,

        label="AIS Prediction"
    )

    ax.axhline(

        regional_predictions[
            "reference_reported_production"
        ].iloc[0],

        color="red",

        linestyle="--",

        label="Reported Reference Production"
    )

    ax.set_xticks(positions)

    ax.set_xticklabels(

        regional_predictions["scenario"],

        rotation=20,

        ha="right"
    )

    ax.set_title(

        f"{region} - AIS Production Predictions",

        fontweight="bold"
    )

    ax.set_ylabel(
        "Estimated Production (Lakh MT)"
    )

    ax.legend()

    ax.grid(alpha=0.3)

fig.suptitle(

    "AIS-Based Sugar Production Scenario Analysis",

    fontsize=18,

    fontweight="bold"
)

save_and_show("ais_prediction_graph.png")

# ============================================================
# 34. AIS FEATURE SENSITIVITY ANALYSIS
# ============================================================

# Measure average relative prediction change when
# each original input feature increases by 1%.

base_predictions = ais_predict(
    X_test
)

feature_sensitivity = {}

for index, feature in enumerate(FEATURES):

    modified_data = X_test.copy()

    modified_data[:, index] *= 1.01

    modified_predictions = ais_predict(
        modified_data
    )

    relative_change = np.mean(

        np.abs(

            modified_predictions -

            base_predictions
        ) /

        np.maximum(

            np.abs(base_predictions),

            1e-8
        )

    ) * 100

    feature_sensitivity[feature] = float(
        relative_change
    )

sensitivity_df = pd.DataFrame({

    "Feature": list(
        feature_sensitivity.keys()
    ),

    "Sensitivity": list(
        feature_sensitivity.values()
    )
})

sensitivity_df = sensitivity_df.sort_values(

    "Sensitivity",

    ascending=True
)

plt.figure(figsize=(11, 6))

bars = plt.barh(

    sensitivity_df["Feature"],

    sensitivity_df["Sensitivity"],

    color="#8e44ad",

    edgecolor="black"
)

for bar in bars:

    value = bar.get_width()

    plt.text(

        value + 0.02,

        bar.get_y() +
        bar.get_height() / 2,

        f"{value:.4f}%",

        va="center"
    )

plt.title(

    "AIS Model Feature Sensitivity Analysis",

    fontsize=16,

    fontweight="bold"
)

plt.xlabel(

    "Mean Absolute Prediction Change (%)\n"
    "For a 1% Increase in Each Feature"
)

plt.xlim(

    0,

    max(

        1.2,

        sensitivity_df["Sensitivity"].max() * 1.20
    )
)

save_and_show(
    "ais_feature_importance_graph.png"
)

# ============================================================
# 35. AIS REGIONAL PERFORMANCE SUMMARY
# ============================================================

regional_summary = result_df.groupby(
    "region"
).agg(

    Average_Reported_Production=(

        "reported_sugar_production",

        "mean"
    ),

    Average_Baseline_Production=(

        "baseline_calculated_production",

        "mean"
    ),

    Average_AIS_Production=(

        "ais_optimized_production",

        "mean"
    ),

    Average_Additional_Production=(

        "additional_production",

        "mean"
    ),

    Average_Improvement_Percentage=(

        "improvement_percentage",

        "mean"
    ),

    Average_Original_Recovery=(

        "original_recovery_rate",

        "mean"
    ),

    Average_AIS_Recovery=(

        "ais_optimized_recovery_rate",

        "mean"
    )
)

regional_summary.round(8).to_csv(

    BASE_DIR / "ais_regional_summary.csv"
)

print("\nAIS Regional Summary:")

print(regional_summary.round(4))

# ============================================================
# 36. ADDITIONAL IMPROVEMENT GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

labels = (

    result_df["region"] +

    "\n" +

    result_df["season"]
)

bars = plt.bar(

    labels,

    result_df["improvement_percentage"],

    color="#27ae60",

    edgecolor="black"
)

for bar in bars:

    height = bar.get_height()

    plt.annotate(

        f"{height:.4f}%",

        (
            bar.get_x() +
            bar.get_width() / 2,

            height
        ),

        xytext=(0, 5),

        textcoords="offset points",

        ha="center",

        fontweight="bold"
    )

plt.title(

    "AIS Hypothetical Sugar Production Improvement",

    fontsize=16,

    fontweight="bold"
)

plt.ylabel("Production Improvement (%)")

plt.margins(y=0.20)

plt.grid(

    axis="y",

    alpha=0.3
)

save_and_show("ais_improvement_graph.png")

# ============================================================
# 37. SAVE YAML CONFIGURATION
# ============================================================

limitations = [

    "The original dataset contains four region-season observations.",

    "Training data is mathematically simulated.",

    "Original observations are not independent external validation.",

    "Tamil Nadu is included in All India figures.",

    "The All India 2017-18 figures are provisional.",

    "AIS regression uses least-squares warm-start initialization.",

    "Estimation score is 100 minus MAPE, not classification accuracy.",

    "Optimization bounds and intervention budget are hypothetical.",

    "The mathematical formula exactly defines simulated production.",

    "AIS cannot establish real-world production improvements without additional data.",

    "Unconstrained comparison scenarios may exceed the AIS intervention budget."
]

config = {

    "project": {

        "name": (
            "AIS-Based Sugar Industry Performance "
            "Analysis and Production Efficiency"
        ),

        "version": "1.0.0",

        "algorithm": "Artificial Immune System",

        "optimization_method": "Clonal Selection Algorithm",

        "created_at": datetime.now().isoformat()
    },

    "dataset": {

        "filename": DATASET_PATH.name,

        "path": str(DATASET_PATH),

        "original_shape": list(raw_df.shape),

        "processed_observations": int(len(df)),

        "simulated_samples": N_SIMULATED_SAMPLES,

        "features": FEATURES,

        "polynomial_features": POLYNOMIAL_FEATURES,

        "target": TARGET,

        "target_unit": "Lakh Metric Tonnes",

        "training_data_type": "Mathematical Simulation"
    },

    "ais_parameters": {

        "population_size": POPULATION_SIZE,

        "generations": GENERATIONS,

        "selected_antibodies": SELECTED_ANTIBODIES,

        "clone_factor": CLONE_FACTOR,

        "mutation_rate": MUTATION_RATE,

        "replacement_rate": REPLACEMENT_RATE,

        "regularization": REGULARIZATION,

        "antibody_dimensions": N_PARAMETERS,

        "initialization": "Least-squares warm start",

        "random_seed": SEED
    },

    "training": {

        "training_samples": int(len(X_train)),

        "validation_samples": int(len(X_val)),

        "testing_samples": int(len(X_test)),

        "original_observations": int(len(X_real)),

        "target_mean": y_mean,

        "target_std": y_std
    },

    "production_optimization": {

        "maximum_factory_change": MAX_FACTORY_CHANGE,

        "maximum_cane_change": MAX_CANE_CHANGE,

        "maximum_recovery_change": MAX_RECOVERY_CHANGE,

        "maximum_recovery_rate": MAX_RECOVERY_RATE,

        "intervention_budget": INTERVENTION_BUDGET,

        "optimization_generations": OPTIMIZATION_GENERATIONS,

        "maximum_relative_cane_per_factory": 1.10,

        "objective": (
            "Maximize formula-based sugar production "
            "under hypothetical intervention and "
            "processing constraints"
        )
    },

    "output_files": {

        "h5": H5_PATH.name,

        "pkl": PKL_PATH.name,

        "yaml": YAML_PATH.name,

        "json": JSON_PATH.name,

        "prefix": PREFIX
    },

    "limitations": limitations
}

with open(

    YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        config,

        file,

        sort_keys=False,

        default_flow_style=False,

        allow_unicode=True
    )

print("\n[SAVED]", YAML_PATH.name)

# ============================================================
# 38. SAVE JSON METRICS
# ============================================================

results = {

    "project_name": config["project"]["name"],

    "algorithm": "Artificial Immune System",

    "generated_at": datetime.now().isoformat(),

    "dataset_information": config["dataset"],

    "ais_parameters": config["ais_parameters"],

    "ais_metrics": ais_metrics,

    "baseline_metrics": baseline_metrics,

    "original_data_comparison": comparison_results,

    "simulation_test_comparison": (
        simulation_comparison_results
    ),

    "best_fitness": float(

        ais_training["best_fitness"]
    ),

    "best_antibody": best_weights.tolist(),

    "training_history": training_history,

    "production_optimization_history": (
        optimization_histories
    ),

    "feature_sensitivity": feature_sensitivity,

    "optimization_results": (

        result_df.to_dict(
            orient="records"
        )
    ),

    "scenario_predictions": (

        prediction_df.to_dict(
            orient="records"
        )
    ),

    "regional_summary": (

        regional_summary.to_dict(
            orient="index"
        )
    ),

    "h5_verification_maximum_difference": (
        maximum_difference
    ),

    "limitations": limitations
}

with open(

    JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        results,

        file,

        indent=4,

        ensure_ascii=False,

        allow_nan=False
    )

print("\n[SAVED]", JSON_PATH.name)

# ============================================================
# 39. VERIFY SAVED PKL MODEL
# ============================================================

loaded_package = joblib.load(
    PKL_PATH
)

sample_input = np.array([

    [37, 81.43, 9.50]

], dtype=float)

sample_poly = polynomial_transform(
    sample_input
)

sample_scaled = loaded_package[
    "scaler"
].transform(sample_poly)

sample_bias = add_bias(
    sample_scaled
)

sample_pkl_prediction = np.maximum(

    (

        sample_bias @

        loaded_package["weights"]
    ) *

    loaded_package["target_std"] +

    loaded_package["target_mean"],

    0.0
)

sample_h5_prediction = loaded_h5_model.predict(

    sample_input.astype(np.float32),

    verbose=0

).flatten()

sample_formula_prediction = (

    sample_input[0, 1] *

    sample_input[0, 2] / 100
)

print("\n" + "=" * 80)
print("SAVED MODEL VERIFICATION")
print("=" * 80)

print("\nSample Input:")

print(

    pd.DataFrame(

        sample_input,

        columns=FEATURES
    )
)

print(

    f"\nMathematical Prediction: "
    f"{sample_formula_prediction:.6f} Lakh MT"
)

print(

    f"AIS PKL Prediction: "
    f"{sample_pkl_prediction[0]:.6f} Lakh MT"
)

print(

    f"AIS H5 Prediction: "
    f"{sample_h5_prediction[0]:.6f} Lakh MT"
)

# ============================================================
# 40. FINAL PERFORMANCE SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL AIS PERFORMANCE SUMMARY")
print("=" * 80)

test_metrics = ais_metrics[
    "simulation_test"
]

original_metrics = ais_metrics[
    "original_observations"
]

print("\nAIS SIMULATION TEST PERFORMANCE")

for name, value in test_metrics.items():

    print(f"{name}: {value:.8f}")

print("\nAIS ORIGINAL DATA PERFORMANCE")

for name, value in original_metrics.items():

    print(f"{name}: {value:.8f}")

print("\nREGIONAL PRODUCTION IMPROVEMENTS")

for region, regional_data in result_df.groupby(
    "region"
):

    average_improvement = (

        regional_data[
            "improvement_percentage"
        ].mean()
    )

    print(

        f"{region}: "
        f"{average_improvement:.4f}% "
        f"average hypothetical improvement"
    )

print("\nBest Model by Original Observation RMSE:")

print(

    comparison_df["RMSE"].idxmin()
)

print("\nBest Model by Original Observation MAPE:")

print(

    comparison_df["MAPE_Percentage"].idxmin()
)

# ============================================================
# 41. VERIFY ALL GENERATED FILES
# ============================================================

expected_files = [

    "ais_sugar_production_model.h5",

    "ais_sugar_production_model.pkl",

    "ais_config.yaml",

    "ais_metrics.json",

    "ais_accuracy_graph.png",

    "ais_heatmap.png",

    "ais_comparison_graph.png",

    "ais_result.csv",

    "ais_result_graph.png",

    "ais_prediction.csv",

    "ais_prediction_graph.png",

    "ais_convergence_graph.png",

    "ais_optimization_graph.png",

    "ais_model_comparison.csv",

    "ais_simulation_comparison.csv",

    "ais_feature_importance_graph.png",

    "ais_regional_summary.csv",

    "ais_improvement_graph.png"
]

print("\n" + "=" * 80)
print("GENERATED AIS OUTPUT FILES")
print("=" * 80)

success_count = 0

for filename in expected_files:

    file_path = BASE_DIR / filename

    if file_path.exists():

        size_kb = (

            file_path.stat().st_size / 1024
        )

        print(

            f"[SUCCESS] {filename} "

            f"({size_kb:.2f} KB)"
        )

        success_count += 1

    else:

        print(

            f"[MISSING] {filename}"
        )

print(

    f"\nTotal Generated Files: "

    f"{success_count}/{len(expected_files)}"
)

print("\nOutput Directory:")

print(BASE_DIR)

print("\n" + "=" * 80)
print("EXPERIMENTAL LIMITATIONS")
print("=" * 80)

for index, limitation in enumerate(

    limitations,

    start=1
):

    print(f"{index}. {limitation}")

print("\n" + "=" * 80)
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)
```
