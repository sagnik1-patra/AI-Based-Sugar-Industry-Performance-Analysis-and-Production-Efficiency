```python
import os
import json
import yaml
import joblib
import random
import warnings
import numpy as np
import pandas as pd
import tensorflow as tf

from pathlib import Path
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import Adam

warnings.filterwarnings("ignore")

# ============================================================
# 1. RANDOM SEED
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

# ============================================================
# 2. FILE PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Sugar Industry Performance Analysis and Production Efficiency"
)

DATASET_PATH = BASE_DIR / (
    "performance_of_sugar_industry_sugar_season_2019.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

H5_PATH = BASE_DIR / "sugar_production_model.h5"
PKL_PATH = BASE_DIR / "sugar_production_model.pkl"
YAML_PATH = BASE_DIR / "config.yaml"
JSON_PATH = BASE_DIR / "metrics.json"

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("=" * 75)
print("AI-BASED SUGAR INDUSTRY PERFORMANCE ANALYSIS")
print("AND PRODUCTION EFFICIENCY")
print("=" * 75)

# ============================================================
# 3. LOAD DATASET
# ============================================================

raw_df = pd.read_csv(DATASET_PATH)

raw_df.columns = raw_df.columns.str.strip()

print("\nOriginal Dataset Shape:", raw_df.shape)

print("\nOriginal Dataset:")
print(raw_df.to_string(index=False))

# ============================================================
# 4. DATA TRANSFORMATION
# ============================================================

# Convert parameter rows into feature columns.

parameter_mapping = {
    "Number of Factories Working":
        "working_factories",

    "Installed Crushing Capacity (in Lakh Tonnes) per day":
        "installed_crushing_capacity",

    "Cane Crushed (in Lakh M.T.)":
        "cane_crushed",

    "Sugar Production from Cane (in Lakh M.T.)":
        "sugar_production",

    "Recovery Rate (in Percentage)":
        "recovery_rate"
}

raw_df["Parameters"] = (
    raw_df["Parameters"].astype(str).str.strip()
)

df = raw_df.set_index("Parameters")

df = df.drop(columns=["S.No"], errors="ignore")

df = df.rename(index=parameter_mapping)

df = df.T.reset_index()

df = df.rename(columns={"index": "region_season"})

df.columns.name = None

required_columns = [
    "working_factories",
    "installed_crushing_capacity",
    "cane_crushed",
    "sugar_production",
    "recovery_rate"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns missing: {missing_columns}"
    )

# ============================================================
# 5. EXTRACT REGION AND YEAR
# ============================================================

df["region"] = np.where(
    df["region_season"].str.contains(
        "Tamil Nadu", case=False
    ),
    "Tamil Nadu",
    "All India"
)

df["season"] = df["region_season"].str.extract(
    r"(\d{4}-\d{2})"
)

df["year"] = pd.to_numeric(
    df["season"].str[:4],
    errors="coerce"
)

df["provisional"] = df["region_season"].str.contains(
    "Provisional",
    case=False
)

df["region_code"] = df["region"].map({
    "Tamil Nadu": 0,
    "All India": 1
})

# ============================================================
# 6. DATA CLEANING
# ============================================================

for column in required_columns:

    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )

print("\nMissing Values:")
print(df.isnull().sum())

# Preserve missing installed capacity.
# Do not invent a value for All India 2017-18.
# Installed capacity is excluded from model training.

df = df.dropna(
    subset=[
        "working_factories",
        "cane_crushed",
        "sugar_production",
        "recovery_rate"
    ]
).reset_index(drop=True)

print("\nProcessed Dataset Shape:", df.shape)

# ============================================================
# 7. FEATURE ENGINEERING
# ============================================================

df["production_per_factory"] = (
    df["sugar_production"] /
    df["working_factories"]
)

df["cane_per_factory"] = (
    df["cane_crushed"] /
    df["working_factories"]
)

df["calculated_sugar_production"] = (
    df["cane_crushed"] *
    df["recovery_rate"] / 100
)

df["production_difference"] = (
    df["sugar_production"] -
    df["calculated_sugar_production"]
)

df["effective_recovery_rate"] = (
    df["sugar_production"] /
    df["cane_crushed"] * 100
)

df["recovery_gap"] = (
    df["effective_recovery_rate"] -
    df["recovery_rate"]
)

print("\nProcessed Dataset:")
print(df.to_string(index=False))

# ============================================================
# 8. INDUSTRY PERFORMANCE ANALYSIS
# ============================================================

print("\n" + "=" * 75)
print("INDUSTRY PERFORMANCE ANALYSIS")
print("=" * 75)

regional_summary = df.groupby("region").agg(
    average_factories=("working_factories", "mean"),
    average_cane_crushed=("cane_crushed", "mean"),
    average_sugar_production=("sugar_production", "mean"),
    average_recovery_rate=("recovery_rate", "mean"),
    average_production_per_factory=(
        "production_per_factory", "mean"
    )
)

print("\nRegional Performance Summary:")
print(regional_summary)

# Calculate percentage changes by region.

regional_changes = {}

for region, group in df.groupby("region"):

    group = group.sort_values("year")

    if len(group) >= 2:

        first = group.iloc[0]
        last = group.iloc[-1]

        regional_changes[region] = {}

        for col in [
            "working_factories",
            "cane_crushed",
            "sugar_production",
            "recovery_rate"
        ]:

            change = (
                (last[col] - first[col]) /
                first[col]
            ) * 100

            regional_changes[region][col] = float(change)

print("\nRegional Percentage Changes:")

for region, values in regional_changes.items():

    print(f"\n{region}:")

    for name, value in values.items():
        print(f"{name}: {value:.2f}%")

# ============================================================
# 9. GENERATE SIMULATION DATA
# ============================================================

# The original dataset contains only four observations.
# Generate hypothetical operating scenarios using:
#
# Sugar Production = Cane Crushed * Recovery Rate / 100
#
# These are simulated observations, not real industry records.

N_SAMPLES = 6000

simulation_records = []

for i in range(N_SAMPLES):

    reference = df.iloc[
        rng.integers(0, len(df))
    ]

    factories = max(
        1,
        int(
            round(
                reference["working_factories"] *
                rng.uniform(0.80, 1.20)
            )
        )
    )

    cane_per_factory = (
        reference["cane_crushed"] /
        reference["working_factories"]
    )

    cane_crushed = (
        factories *
        cane_per_factory *
        rng.uniform(0.70, 1.30)
    )

    recovery_rate = np.clip(
        reference["recovery_rate"] *
        rng.uniform(0.85, 1.15),
        6.0,
        13.0
    )

    sugar_production = (
        cane_crushed * recovery_rate / 100
    )

    simulation_records.append({
        "working_factories": factories,
        "cane_crushed": cane_crushed,
        "recovery_rate": recovery_rate,
        "sugar_production": sugar_production
    })

simulated_df = pd.DataFrame(simulation_records)

print("\nSimulated Dataset Shape:", simulated_df.shape)

print("\nSimulation Sample:")
print(simulated_df.head())

# ============================================================
# 10. FEATURE AND TARGET SELECTION
# ============================================================

FEATURES = [
    "working_factories",
    "cane_crushed",
    "recovery_rate"
]

TARGET = "sugar_production"

X = simulated_df[FEATURES].astype("float32")

y = simulated_df[TARGET].astype("float32")

X_real = df[FEATURES].astype("float32")

y_real = df[TARGET].astype("float32")

print("\nSelected Features:", FEATURES)
print("Target Variable:", TARGET)

# ============================================================
# 11. TRAIN, VALIDATION AND TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

print("\nTraining Samples:", len(X_train))
print("Validation Samples:", len(X_val))
print("Testing Samples:", len(X_test))
print("Original Observations:", len(X_real))

# ============================================================
# 12. RANDOM FOREST REGRESSION MODEL
# ============================================================

print("\n" + "=" * 75)
print("TRAINING RANDOM FOREST REGRESSION MODEL")
print("=" * 75)

rf_pipeline = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=200,
            max_depth=18,
            min_samples_split=2,
            min_samples_leaf=1,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

rf_pipeline.fit(
    X_train_full,
    y_train_full
)

rf_predictions = rf_pipeline.predict(X_test)

rf_real_predictions = rf_pipeline.predict(X_real)

print("\nRandom Forest Training Completed.")

# ============================================================
# 13. ARTIFICIAL NEURAL NETWORK MODEL
# ============================================================

print("\n" + "=" * 75)
print("TRAINING ARTIFICIAL NEURAL NETWORK")
print("=" * 75)

# Normalization is embedded inside the model so that
# the saved H5 file accepts original, unscaled inputs.

normalizer = tf.keras.layers.Normalization(
    axis=-1,
    name="input_normalization"
)

normalizer.adapt(
    X_train.to_numpy(dtype=np.float32)
)

# Scale the target during training and restore its
# original units using the final Rescaling layer.

target_mean = float(y_train.mean())

target_std = float(y_train.std(ddof=0))

if target_std == 0:
    target_std = 1.0

ann_model = Sequential([
    Input(shape=(len(FEATURES),)),

    normalizer,

    Dense(128, activation="relu"),

    Dense(64, activation="relu"),

    Dense(32, activation="relu"),

    Dense(16, activation="relu"),

    Dense(1, activation="linear"),

    tf.keras.layers.Rescaling(
        scale=target_std,
        offset=target_mean,
        name="output_rescaling"
    )
])

ann_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = ann_model.fit(
    X_train.to_numpy(dtype=np.float32),
    y_train.to_numpy(dtype=np.float32),

    validation_data=(
        X_val.to_numpy(dtype=np.float32),
        y_val.to_numpy(dtype=np.float32)
    ),

    epochs=200,
    batch_size=32,

    callbacks=[early_stopping],

    verbose=1
)

ann_predictions = ann_model.predict(
    X_test.to_numpy(dtype=np.float32),
    verbose=0
).flatten()

ann_real_predictions = ann_model.predict(
    X_real.to_numpy(dtype=np.float32),
    verbose=0
).flatten()

print("\nANN Training Completed.")

# ============================================================
# 14. MATHEMATICAL BASELINE
# ============================================================

baseline_predictions = (
    X_test["cane_crushed"] *
    X_test["recovery_rate"] / 100
).to_numpy()

baseline_real_predictions = (
    X_real["cane_crushed"] *
    X_real["recovery_rate"] / 100
).to_numpy()

# ============================================================
# 15. MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_model(actual, predicted):

    actual = np.asarray(actual, dtype=float)

    predicted = np.asarray(
        predicted,
        dtype=float
    ).flatten()

    mae = mean_absolute_error(
        actual, predicted
    )

    mse = mean_squared_error(
        actual, predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual, predicted
    )

    nonzero_mask = actual != 0

    if np.any(nonzero_mask):

        mape = np.mean(
            np.abs(
                (
                    actual[nonzero_mask] -
                    predicted[nonzero_mask]
                ) /
                actual[nonzero_mask]
            )
        ) * 100

        estimation_score = 100 - mape

    else:

        mape = None
        estimation_score = None

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2_Score": float(r2),

        "MAPE_Percentage": (
            float(mape)
            if mape is not None
            else None
        ),

        "Estimation_Score_Percentage": (
            float(estimation_score)
            if estimation_score is not None
            else None
        )
    }

# ============================================================
# 16. SIMULATION TEST EVALUATION
# ============================================================

simulation_metrics = {

    "Mathematical_Baseline": evaluate_model(
        y_test,
        baseline_predictions
    ),

    "Random_Forest": evaluate_model(
        y_test,
        rf_predictions
    ),

    "Artificial_Neural_Network": evaluate_model(
        y_test,
        ann_predictions
    )
}

# ============================================================
# 17. ORIGINAL DATA EVALUATION
# ============================================================

real_data_metrics = {

    "Mathematical_Baseline": evaluate_model(
        y_real,
        baseline_real_predictions
    ),

    "Random_Forest": evaluate_model(
        y_real,
        rf_real_predictions
    ),

    "Artificial_Neural_Network": evaluate_model(
        y_real,
        ann_real_predictions
    )
}

print("\n" + "=" * 75)
print("SIMULATION TEST PERFORMANCE")
print("=" * 75)

simulation_comparison = pd.DataFrame(
    simulation_metrics
).T

print(simulation_comparison.round(4))

print("\n" + "=" * 75)
print("ORIGINAL DATA PERFORMANCE")
print("=" * 75)

real_comparison = pd.DataFrame(
    real_data_metrics
).T

print(real_comparison.round(4))

# ============================================================
# 18. FEATURE IMPORTANCE
# ============================================================

rf_model = rf_pipeline.named_steps["model"]

feature_importance = {
    feature: float(importance)

    for feature, importance in zip(
        FEATURES,
        rf_model.feature_importances_
    )
}

print("\nRandom Forest Feature Importance:")

for feature, importance in feature_importance.items():

    print(f"{feature}: {importance:.4f}")

# ============================================================
# 19. PRODUCTION EFFICIENCY ANALYSIS
# ============================================================

efficiency_results = []

for _, row in df.iterrows():

    current_production = float(
        row["sugar_production"]
    )

    current_recovery = float(
        row["recovery_rate"]
    )

    cane_crushed = float(
        row["cane_crushed"]
    )

    # Hypothetical 5% relative recovery improvement.

    improved_recovery = min(
        current_recovery * 1.05,
        13.0
    )

    baseline_estimated_production = (
        cane_crushed *
        current_recovery / 100
    )

    optimized_production = (
        cane_crushed *
        improved_recovery / 100
    )

    additional_production = (
        optimized_production -
        baseline_estimated_production
    )

    improvement_percentage = (
        additional_production /
        baseline_estimated_production
    ) * 100

    efficiency_results.append({

        "Region": str(row["region"]),

        "Season": str(row["season"]),

        "Current_Production": current_production,

        "Current_Recovery_Rate": current_recovery,

        "Improved_Recovery_Rate": float(
            improved_recovery
        ),

        "Baseline_Estimated_Production": float(
            baseline_estimated_production
        ),

        "Optimized_Production": float(
            optimized_production
        ),

        "Additional_Production": float(
            additional_production
        ),

        "Improvement_Percentage": float(
            improvement_percentage
        )
    })

efficiency_df = pd.DataFrame(
    efficiency_results
)

print("\n" + "=" * 75)
print("HYPOTHETICAL PRODUCTION EFFICIENCY ANALYSIS")
print("=" * 75)

print(efficiency_df.round(4).to_string(index=False))

# ============================================================
# 20. SAVE H5 MODEL
# ============================================================

ann_model.save(
    str(H5_PATH),
    include_optimizer=False
)

print("\nH5 Model Saved:")
print(H5_PATH)

# ============================================================
# 21. SAVE PKL MODEL
# ============================================================

joblib.dump(
    rf_pipeline,
    PKL_PATH
)

print("\nPKL Model Saved:")
print(PKL_PATH)

# ============================================================
# 22. CREATE YAML CONFIGURATION
# ============================================================

config = {

    "project": {
        "name": (
            "AI-Based Sugar Industry Performance "
            "Analysis and Production Efficiency"
        ),

        "version": "1.0.0",

        "created_at": datetime.now().isoformat(),

        "domain": "Agricultural and Industrial Analytics",

        "task": "Regression and Efficiency Analysis"
    },

    "dataset": {

        "filename": DATASET_PATH.name,

        "path": str(DATASET_PATH),

        "original_rows": int(raw_df.shape[0]),

        "original_columns": int(raw_df.shape[1]),

        "processed_observations": int(len(df)),

        "simulation_samples": int(N_SAMPLES),

        "regions": df["region"].unique().tolist(),

        "seasons": df["season"].unique().tolist(),

        "features": FEATURES,

        "target": TARGET,

        "target_unit": "Lakh Metric Tonnes",

        "training_data_type": "Mathematical Simulation",

        "missing_capacity_handling": (
            "Preserved as missing and excluded from models"
        ),

        "simulation_formula": (
            "sugar_production = "
            "cane_crushed * recovery_rate / 100"
        )
    },

    "training": {

        "random_seed": SEED,

        "training_samples": int(len(X_train)),

        "validation_samples": int(len(X_val)),

        "testing_samples": int(len(X_test)),

        "real_observations": int(len(X_real)),

        "epochs": 200,

        "batch_size": 32,

        "learning_rate": 0.001,

        "early_stopping_patience": 20
    },

    "random_forest": {

        "n_estimators": 200,

        "max_depth": 18,

        "min_samples_split": 2,

        "min_samples_leaf": 1,

        "scaling": "StandardScaler"
    },

    "neural_network": {

        "input_features": len(FEATURES),

        "hidden_layers": [128, 64, 32, 16],

        "activation": "relu",

        "output_activation": "linear",

        "optimizer": "Adam",

        "loss": "Mean Squared Error",

        "normalization": "Embedded Keras Normalization",

        "target_mean": target_mean,

        "target_std": target_std
    },

    "efficiency_analysis": {

        "relative_recovery_improvement": 0.05,

        "maximum_simulated_recovery_rate": 13.0,

        "type": "Hypothetical Scenario"
    },

    "output_files": {

        "h5_model": str(H5_PATH),

        "pkl_model": str(PKL_PATH),

        "yaml_configuration": str(YAML_PATH),

        "json_metrics": str(JSON_PATH)
    },

    "limitations": [

        "Only four original region-season observations.",

        "Models are trained on simulated data.",

        "Tamil Nadu is included in All India totals.",

        "All India 2017-18 figures are provisional.",

        "Original data evaluation is not independent validation.",

        "Estimation score is 100 minus MAPE, not classification accuracy.",

        "Mathematical baseline is sufficient for the simulated target."
    ]
}

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        default_flow_style=False,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML Configuration Saved:")
print(YAML_PATH)

# ============================================================
# 23. CREATE JSON RESULTS
# ============================================================

real_predictions = []

for i in range(len(df)):

    real_predictions.append({

        "Region": str(df.iloc[i]["region"]),

        "Season": str(df.iloc[i]["season"]),

        "Actual_Production": float(y_real.iloc[i]),

        "Baseline_Prediction": float(
            baseline_real_predictions[i]
        ),

        "Random_Forest_Prediction": float(
            rf_real_predictions[i]
        ),

        "ANN_Prediction": float(
            ann_real_predictions[i]
        )
    })

results = {

    "project_name": config["project"]["name"],

    "generated_at": datetime.now().isoformat(),

    "dataset_information": {

        "original_shape": list(raw_df.shape),

        "processed_shape": list(df.shape),

        "simulation_shape": list(simulated_df.shape),

        "features": FEATURES,

        "target": TARGET
    },

    "simulation_test_metrics": simulation_metrics,

    "original_data_metrics": real_data_metrics,

    "feature_importance": feature_importance,

    "regional_summary": (
        regional_summary.to_dict(orient="index")
    ),

    "regional_percentage_changes": regional_changes,

    "original_data_predictions": real_predictions,

    "efficiency_analysis": efficiency_results,

    "training_history": {

        key: [float(value) for value in values]

        for key, values in history.history.items()
    },

    "limitations": config["limitations"]
}

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        results,
        file,
        indent=4,
        ensure_ascii=False,
        allow_nan=False
    )

print("\nJSON Metrics Saved:")
print(JSON_PATH)

# ============================================================
# 24. LOAD AND VERIFY SAVED MODELS
# ============================================================

print("\n" + "=" * 75)
print("VERIFYING SAVED MODELS")
print("=" * 75)

loaded_rf_model = joblib.load(
    PKL_PATH
)

loaded_ann_model = tf.keras.models.load_model(
    str(H5_PATH),
    compile=False
)

# Sample input representing a hypothetical
# Tamil Nadu sugar production scenario.

sample_input = pd.DataFrame([{

    "working_factories": 37,

    "cane_crushed": 81.43,

    "recovery_rate": 9.50

}])[FEATURES]

rf_sample_prediction = loaded_rf_model.predict(
    sample_input
)[0]

ann_sample_prediction = loaded_ann_model.predict(
    sample_input.to_numpy(dtype=np.float32),
    verbose=0
).flatten()[0]

baseline_sample_prediction = (
    81.43 * 9.50 / 100
)

print("\nSample Input:")
print(sample_input.to_string(index=False))

print(
    f"\nMathematical Baseline: "
    f"{baseline_sample_prediction:.4f} Lakh MT"
)

print(
    f"Random Forest Prediction: "
    f"{rf_sample_prediction:.4f} Lakh MT"
)

print(
    f"ANN Prediction: "
    f"{ann_sample_prediction:.4f} Lakh MT"
)

# ============================================================
# 25. FINAL OUTPUT VERIFICATION
# ============================================================

print("\n" + "=" * 75)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 75)

output_files = [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]

for file_path in output_files:

    if file_path.exists():

        size_kb = (
            file_path.stat().st_size / 1024
        )

        print(
            f"[SUCCESS] {file_path.name} "
            f"({size_kb:.2f} KB)"
        )

    else:

        print(
            f"[FAILED] {file_path.name}"
        )

print("\nAll output files are saved in:")
print(BASE_DIR)

print("\nNOTE:")
print(
    "Models were trained using simulated observations."
)

print(
    "Results are intended for experimental and "
    "educational analysis."
)

print("=" * 75)
```
